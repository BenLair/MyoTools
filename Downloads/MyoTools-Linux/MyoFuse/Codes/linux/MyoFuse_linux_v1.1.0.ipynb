```python
## IMPORT LIBRARIES ##

import os
from pathlib import Path
import json
import numpy as np
import torch
import pandas as pd
from cellpose import models
from aicsimageio import AICSImage
from tifffile import imwrite 
from skimage.io import imread, imsave
from skimage.measure import label
from skimage.morphology import dilation, disk
from skimage.transform import resize
from skimage.util import img_as_ubyte
from skimage.segmentation import find_boundaries
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms as transforms
from scipy import ndimage
import warnings
import asyncio
import html
import io
import colorsys
import traceback
from PIL import Image as PILImage, ImageDraw

import ipywidgets as widgets
from ipyfilechooser import FileChooser

def default_model_file(name):
    """Use a model downloaded by MyoTools, when it is present."""
    import os
    folder = Path(os.environ.get("MYOTOOLS_MODEL_DIR", str(Path.cwd() / "Hub" / "runtime" / "models")))
    model = folder / name
    return model if model.is_file() else None


def default_image_directory():
    """Use the user's real Desktop, or their home if no Desktop is available.

    This standard-library-only function is embedded in the notebooks so they
    also work when launched directly, without importing or starting the hub.
    """
    import os
    import subprocess
    from pathlib import Path

    if os.name == "nt":
        command = ["powershell.exe", "-NoProfile", "-Command",
                   "[Console]::OutputEncoding = [System.Text.Encoding]::UTF8; [Environment]::GetFolderPath('DesktopDirectory')"]
    else:
        command = ["xdg-user-dir", "DESKTOP"]
    try:
        result = subprocess.run(command, capture_output=True, text=True, timeout=5,
                                encoding="utf-8", errors="replace")
        if result.returncode == 0 and result.stdout.strip():
            desktop = Path(result.stdout.strip().lstrip("\ufeff")).expanduser()
            if desktop.is_absolute() and desktop.is_dir():
                return desktop
    except (OSError, subprocess.SubprocessError):
        pass
    home = Path.home()
    return next((path for path in (home / "Desktop", home / "Bureau") if path.is_dir()), home)

from IPython.display import display, Markdown, HTML

if torch.cuda.is_available() is True:
    try:
        import cupy as cu
        from cucim.skimage.morphology import dilation, disk
        cuda = True
    except ImportError:
        from skimage.morphology import dilation, disk
        cuda = False
else:
    from skimage.morphology import dilation, disk
    cuda = False

warnings.filterwarnings('ignore')

#### FUNCTIONS ####

def load_checkpoint(filename):
    if os.path.exists(filename):
        with open(filename, 'r') as f:
            return json.load(f)
    return {}

def save_checkpoint(filename, data):
    temporary = filename + ".tmp"
    with open(temporary, "w") as handle:
        json.dump(data, handle, indent=2)
    os.replace(temporary, filename)

def reset_checkpoints(parent_directory_value):
    if not parent_directory_value or not os.path.isdir(parent_directory_value):
        raise ValueError("Select an existing image folder before launching.")
    log_folder = os.path.join(parent_directory_value, "Log")
    os.makedirs(log_folder, exist_ok=True)
    result_file = os.path.join(parent_directory_value, "Fusion_index.xlsx")
    if os.path.isfile(result_file):
        import shutil
        from datetime import datetime
        stamp = datetime.now().strftime("%Y%m%d_%H%M%S_%f")
        backup = os.path.join(log_folder, f"Fusion_index_before_launch_{stamp}.xlsx")
        shutil.copy2(result_file, backup)
        os.remove(result_file)
        print(f"Previous results backed up to: {backup}")
    for name in ("channel_split_checkpoint.json", "segmentation_checkpoint.json",
                 "classification_checkpoint.json", "classification_results.json"):
        path = os.path.join(log_folder, name)
        if os.path.exists(path):
            os.remove(path)

def normalize_image(image, percentile_max, nbins=512, smoothing=2):
    
     # Extract non-0 pixels
    nonzero_pixels = image[image > 0]
    if nonzero_pixels.size == 0:
        return image 

    # Compute CDF
    hist, bin_edges = np.histogram(nonzero_pixels, bins=nbins)
    cdf = np.cumsum(hist)
    cdf = cdf / cdf[-1]  # normalize between 0 and 1
    diff = np.diff(cdf)  # size nbins-1
    secdiff = np.diff(diff)  # size nbins-2

    # Smoothing of second derivative
    if smoothing > 1:
        kernel = np.ones(smoothing) / smoothing
        secdiff_smooth = np.convolve(secdiff, kernel, mode='same')
    else:
        secdiff_smooth = secdiff

    # Search for maximum
    idx_knee = np.argmax(secdiff_smooth)
    idx_knee_in_cdf = idx_knee + 1 

    if idx_knee_in_cdf < 0 or idx_knee_in_cdf >= len(bin_edges):
        idx_knee_in_cdf = 0  # fallback if out of boundaries

    # Set new min threshold
    new_min = bin_edges[idx_knee_in_cdf]

    # Set high threshold
    max_val = np.percentile(nonzero_pixels, percentile_max)

    # Normalize between 0 and 1
    norm_image = np.zeros_like(image, dtype=np.float32)

    image_clipped = np.clip(image, new_min, max_val)
    denom = (max_val - new_min) if (max_val > new_min) else 1e-9

    mask = (image_clipped >= new_min)
    norm_image[mask] = (image_clipped[mask] - new_min) / denom

    norm_image = np.clip(norm_image, 0, 1)
    
    return norm_image
def max_to_one(im):
    
    im = im / im.max()

    return im
    
## Collect label centroids

def compute_label_centroids(label_image):
    
    label_image = label_image.astype(np.int32, copy=False)
    labels_unique = np.unique(label_image)
    labels_unique = labels_unique[labels_unique != 0]

    weights = np.ones_like(label_image, dtype=np.float32)
    cm = ndimage.center_of_mass(weights, labels=label_image, index=labels_unique)

    return labels_unique, cm

## Prediction functions

class PredictionDataset(Dataset):
    """
      prop.centroid = (row, col)  =>  cx = row, cy = col  
    """

    def __init__(self, image, labels, label_ids, centroids, half_patch_size, device, config_dict):
        super().__init__()
        self.image = image
        self.labels = labels
        self.label_ids = label_ids
        self.centroids = centroids
        self.half_patch_size = half_patch_size
        self.device = device
        self.config_dict = config_dict

        self.transform = transforms.Compose([
            transforms.ToTensor()
        ])

        pad = half_patch_size + 1

        self.image = np.pad(
            self.image,
            ((pad, pad), (pad, pad), (0, 0)),
            mode="constant"
        )
        if self.labels.dtype != np.int32:
            self.labels = self.labels.astype(np.int32, copy=False)
        self.labels = np.pad(
            self.labels,
            ((pad, pad), (pad, pad)),
            mode="constant"
        )

    def __len__(self):
        return len(self.label_ids)

    def __getitem__(self, idx):
        try:
           
            row, col = self.centroids[idx] 
            cx = int(row)
            cy = int(col)
            cx += self.half_patch_size + 1
            cy += self.half_patch_size + 1
            hps = self.half_patch_size
            xmin, xmax = cx - hps, cx + hps
            ymin, ymax = cy - hps, cy + hps

            # Extract patch
            patch_img = self.image[xmin:xmax, ymin:ymax, :].copy()
            patch_mask = self.labels[xmin:xmax, ymin:ymax].copy()

            label_id = self.label_ids[idx]

            # Normalize
            patch_img = max_to_one(patch_img)

            # Binarize
            patch_mask[patch_mask != label_id] = 0
            patch_mask[patch_mask == label_id] = 1

            # Dilation if stated in the config file
            do_dilate = json.loads(self.config_dict["options"]["dilation"]["dilate_mask"].lower())
            if do_dilate:
                se_size = int(self.config_dict["options"]["dilation"]["str_element_size"])
                strel = disk(se_size)
                patch_mask = dilation(patch_mask, strel)
                patch_img *= patch_mask[..., None]

            # Mask concatenation as 4th channel
            out = np.zeros((patch_img.shape[0], patch_img.shape[1], 4), dtype=np.float32)
            out[..., :3] = patch_img
            out[..., 3] = patch_mask

            # Transformation to (C, H, W) + Tensor
            out = self.transform(out)
            return out.to(self.device)

        except Exception as e:
            print(f"[Dataset] Erreur index {idx} : {e}")
            return None
pass

## Save prediction image

def save_colored_predictions_downsample(
    labels, 
    predictions, 
    used_labels, 
    myotube_image, 
    output_path, 
    factor=1
):
    if myotube_image.ndim == 3 and myotube_image.shape[2] == 3:
        myotube_image = np.mean(myotube_image, axis=2)
    H, W = myotube_image.shape

    newH, newW = H//factor, W//factor
    myotube_ds = resize(myotube_image, (newH, newW),
                        preserve_range=True,
                        anti_aliasing=True)

    myotube_rgb = np.stack([myotube_ds]*3, axis=-1)

    boundaries = find_boundaries(labels, mode='inner')
    boundary_labels = labels.copy()
    boundary_labels[~boundaries] = 0

    if len(predictions) != len(used_labels):
        raise ValueError("Number of predictions != Number of labels.")
    label_to_pred = dict(zip(used_labels, predictions))

    color_0 = [1.0, 0.0, 0.0]  # Rouge = Nuclei Out
    color_1 = [0.0, 1.0, 0.0]  # Vert = Nuclei In

    coords = np.column_stack(np.nonzero(boundary_labels))
    for (y, x) in coords:
        lb = boundary_labels[y, x]
        pred_class = label_to_pred.get(lb, None)
        if pred_class is not None:
            yd = y // factor
            xd = x // factor
            if yd < newH and xd < newW:
                if pred_class == 0:
                    myotube_rgb[yd, xd] = color_0
                else:
                    myotube_rgb[yd, xd] = color_1
    myotube_rgb_8 = img_as_ubyte(myotube_rgb)
    imsave(output_path, myotube_rgb_8)
    pass

#### ANALYSIS ####

import hashlib


def file_stamp(path):
    stat = os.stat(path)
    return [os.path.abspath(path), stat.st_size, stat.st_mtime_ns]


def stage_signature(paths, parameters):
    payload = {
        "inputs": [file_stamp(path) for path in paths],
        "parameters": parameters,
    }
    encoded = json.dumps(payload, sort_keys=True).encode("utf-8")
    return hashlib.sha256(encoded).hexdigest()


def outputs_exist(paths):
    return all(os.path.isfile(path) and os.path.getsize(path) > 0 for path in paths)


def step_is_complete(checkpoint, key, signature, outputs):
    return checkpoint.get("records", {}).get(key) == signature and outputs_exist(outputs)


def mark_complete(checkpoint_path, checkpoint, key, signature):
    checkpoint.setdefault("records", {})[key] = signature
    save_checkpoint(checkpoint_path, checkpoint)


def image_stem(filename):
    return os.path.splitext(filename)[0]


def read_channel_pair(path, myotube_channel, nuclei_channel):
    if min(myotube_channel, nuclei_channel) < 0 or myotube_channel == nuclei_channel:
        raise ValueError("Choose two distinct, non-negative channel indices.")
    image = AICSImage(path, reconstruct_mosaic=False)
    highest_channel = max(myotube_channel, nuclei_channel)
    if image.dims.C > highest_channel:
        channels = image.get_image_data("CYX", T=0, Z=0)
    elif path.lower().endswith((".tif", ".tiff")):
        # Some plain TIFF readers expose the channels as samples rather than C.
        raw = imread(path)
        if raw.ndim == 3 and raw.shape[-1] > highest_channel and raw.shape[-1] <= 4:
            channels = np.moveaxis(raw, -1, 0)
        elif raw.ndim == 3 and raw.shape[0] > highest_channel and raw.shape[0] <= 4:
            channels = raw
        else:
            raise ValueError(f"Cannot identify the requested channels; dimensions: {image.dims}.")
    else:
        raise ValueError(f"Requested channel is absent; dimensions: {image.dims}.")
    myotube = channels[myotube_channel]
    nuclei = channels[nuclei_channel]
    if myotube.ndim != 2 or nuclei.shape != myotube.shape:
        raise ValueError("Both selected channels must be 2D images of the same shape.")
    return myotube, nuclei


def run_entire_analysis(
    myotube_channel, nuclei_channel, dia, percentile_max,
    half_patch_size, batch_size, split_images, run_segmentation, seg_device,
    run_classification, class_device, save_normalization, save_prediction,
    downsampling_factor, parent_directory_value, seg_directory_value,
    class_directory_value, emit=None
):
    emit = emit or (lambda event: None)
    if not parent_directory_value or not os.path.isdir(parent_directory_value):
        raise ValueError("Select an existing image folder.")
    if dia <= 0 or batch_size < 1 or half_patch_size < 1 or downsampling_factor < 1:
        raise ValueError("Batch size, half patch size and downsampling factor must be positive.")

    myotube_folder = os.path.join(parent_directory_value, "Images")
    nuclei_folder = os.path.join(parent_directory_value, "Nuclei")
    masks_folder = os.path.join(parent_directory_value, "Masks")
    predictions_dir = os.path.join(parent_directory_value, "Predictions")
    norm_dir = os.path.join(parent_directory_value, "Normalized_images")
    log_folder = os.path.join(parent_directory_value, "Log")
    os.makedirs(log_folder, exist_ok=True)
    os.makedirs(myotube_folder, exist_ok=True)
    os.makedirs(nuclei_folder, exist_ok=True)
    os.makedirs(masks_folder, exist_ok=True)

    split_checkpoint_path = os.path.join(log_folder, "channel_split_checkpoint.json")
    seg_checkpoint_path = os.path.join(log_folder, "segmentation_checkpoint.json")
    class_checkpoint_path = os.path.join(log_folder, "classification_checkpoint.json")
    result_store_path = os.path.join(log_folder, "classification_results.json")

    if split_images:
        display(Markdown("## CHANNEL SPLITTING ##"))
        sources = sorted(f for f in os.listdir(parent_directory_value)
                         if f.lower().endswith((".tif", ".tiff", ".czi")))
        stems = [image_stem(f) for f in sources]
        if len(set(stems)) != len(stems):
            raise ValueError("Two input images have the same name without their extension.")
        split_checkpoint = load_checkpoint(split_checkpoint_path)
        successful_stems = set()
        emit({"kind":"stage","stage":"splitting","total":len(sources)})
        for index, filename in enumerate(sources, 1):
            outcome="completed"
            emit({"kind":"file","filename":filename,"state":"running","current":index-1,"total":len(sources)})
            source = os.path.join(parent_directory_value, filename)
            stem = image_stem(filename)
            image_out = os.path.join(myotube_folder, stem + ".tif")
            nuclei_out = os.path.join(nuclei_folder, stem + ".tif")
            signature = stage_signature([source], [myotube_channel, nuclei_channel])
            try:
                if not step_is_complete(split_checkpoint, filename, signature,
                                        [image_out, nuclei_out]):
                    myotube, nuclei = read_channel_pair(
                        source, myotube_channel, nuclei_channel
                    )
                    imsave(image_out, myotube.astype(np.uint16))
                    imsave(nuclei_out, nuclei.astype(np.uint16))
                    mark_complete(split_checkpoint_path, split_checkpoint, filename,
                                  signature)
                successful_stems.add(stem)
            except Exception as error:
                split_checkpoint.get("records", {}).pop(filename, None)
                save_checkpoint(split_checkpoint_path, split_checkpoint)
                outcome="failed"
                print(f"Failed channel splitting for {filename}: {error}")
            finally:
                emit({"kind":"file","filename":filename,"state":outcome,"current":index,"total":len(sources)})
    else:
        successful_stems = None

    if run_segmentation:
        if not seg_directory_value or not os.path.isfile(seg_directory_value):
            raise ValueError("Select an existing segmentation model file.")
        display(Markdown("## SEGMENTATION ##"))
        seg_checkpoint = load_checkpoint(seg_checkpoint_path)
        nuclei_files = sorted(f for f in os.listdir(nuclei_folder)
                              if f.lower().endswith((".tif", ".tiff"))
                              and (successful_stems is None or image_stem(f) in successful_stems))
        emit({"kind":"stage","stage":"segmentation","total":len(nuclei_files)})
        seg_model = None
        segmented_stems = set()
        for index, filename in enumerate(nuclei_files, 1):
            outcome="completed"
            emit({"kind":"file","filename":filename,"state":"running","current":index-1,"total":len(nuclei_files)})
            source = os.path.join(nuclei_folder, filename)
            mask_out = os.path.join(masks_folder, filename)
            signature = stage_signature([source, seg_directory_value], [dia])
            try:
                if not step_is_complete(seg_checkpoint, filename, signature, [mask_out]):
                    if seg_model is None:
                        seg_model = models.CellposeModel(
                            gpu=seg_device == "GPU", pretrained_model=seg_directory_value
                        )
                    image = AICSImage(source).get_image_data("YX", T=0, C=0, Z=0)
                    masks, _flows, _styles = seg_model.eval(
                        image, diameter=dia, channels=[0, 0], normalize=True
                    )
                    # An all-zero mask records an image with no detected nuclei.
                    imwrite(mask_out, masks.astype(np.uint32))
                    mark_complete(seg_checkpoint_path, seg_checkpoint, filename,
                                  signature)
                segmented_stems.add(image_stem(filename))
            except Exception as error:
                seg_checkpoint.get("records", {}).pop(filename, None)
                save_checkpoint(seg_checkpoint_path, seg_checkpoint)
                outcome="failed"
                print(f"Failed segmentation for {filename}: {error}")
            finally:
                emit({"kind":"file","filename":filename,"state":outcome,"current":index,"total":len(nuclei_files)})
    else:
        segmented_stems = None

    if not run_classification:
        return
    if not class_directory_value or not os.path.isfile(class_directory_value):
        raise ValueError("Select an existing classification model file.")
    config_path = os.path.join(os.path.dirname(class_directory_value), "Config.json")
    if not os.path.isfile(config_path):
        raise ValueError(f"Missing classifier configuration: {config_path}")

    display(Markdown("## CLASSIFICATION ##"))
    class_checkpoint = load_checkpoint(class_checkpoint_path)
    previous_workbook = os.path.join(parent_directory_value, "Fusion_index.xlsx")
    if not os.path.isfile(result_store_path) and os.path.isfile(previous_workbook):
        # Old versions have no per-image results or parameter signatures.
        # Keep the old workbook while rebuilding a verifiable result store.
        import shutil
        from datetime import datetime
        stamp = datetime.now().strftime("%Y%m%d_%H%M%S_%f")
        backup = os.path.join(log_folder, f"Fusion_index_before_migration_{stamp}.xlsx")
        shutil.copy2(previous_workbook, backup)
        print("Previous workbook backed up. Unverified legacy images will be "
              "processed once to establish resumable results.")
    results = load_checkpoint(result_store_path)
    myotube_files = sorted(f for f in os.listdir(myotube_folder)
                           if f.lower().endswith((".tif", ".tiff"))
                           and (successful_stems is None or image_stem(f) in successful_stems))
    if len({image_stem(f) for f in myotube_files}) != len(myotube_files):
        raise ValueError("Two myotube images have the same name without their extension.")

    emit({"kind":"stage","stage":"classification","total":len(myotube_files)})
    device = torch.device("cuda" if class_device == "GPU" and torch.cuda.is_available()
                          else "cpu")
    model = None
    valid_rows = []
    for index, filename in enumerate(myotube_files, 1):
        outcome="completed"
        emit({"kind":"file","filename":filename,"state":"running","current":index-1,"total":len(myotube_files)})
        stem = image_stem(filename)
        img_path = os.path.join(myotube_folder, filename)
        mask_path = os.path.join(masks_folder, filename)
        prediction_out = os.path.join(predictions_dir, stem + "_prediction.tif")
        norm_out = os.path.join(norm_dir, filename)
        expected_outputs = ([prediction_out] if save_prediction else []) + (
            [norm_out] if save_normalization else []
        )
        try:
            if segmented_stems is not None and stem not in segmented_stems:
                outcome="skipped"
                print(f"Skipping {filename}: segmentation did not complete.")
                continue
            if not os.path.isfile(mask_path):
                outcome="skipped"
                print(f"Skipping {filename}: no matching mask.")
                continue

            signature = stage_signature(
                [img_path, mask_path, class_directory_value, config_path],
                [percentile_max, half_patch_size, save_prediction,
                 downsampling_factor if save_prediction else None, save_normalization]
            )
            stored = results.get(filename)
            if (stored is not None and stored.get("signature") == signature
                    and outputs_exist(expected_outputs)):
                valid_rows.append(stored["row"])
                if not step_is_complete(class_checkpoint, filename, signature,
                                        expected_outputs):
                    mark_complete(class_checkpoint_path, class_checkpoint,
                                  filename, signature)
                outcome="reused"
                print(f"{filename} already classified.")
                continue

            if model is None:
                with open(config_path, "r") as handle:
                    config_dict = json.load(handle)
                checkpoint_model = torch.load(
                    class_directory_value, map_location=device, weights_only=False
                )
                model = checkpoint_model["model"].to(device)
                model.eval()
            image_data = imread(img_path).astype(np.float32)
            if image_data.ndim == 2:
                image_data = np.stack([image_data] * 3, axis=-1)
            image_data = normalize_image(image_data, percentile_max)
            if save_normalization:
                os.makedirs(norm_dir, exist_ok=True)
                imsave(norm_out, img_as_ubyte(image_data))

            mask_data = imread(mask_path)
            label_ids, centroids = compute_label_centroids(mask_data)
            image_preds = []
            if len(label_ids):
                dataset = PredictionDataset(
                    image=image_data, labels=mask_data, label_ids=label_ids,
                    centroids=centroids, half_patch_size=half_patch_size,
                    device=device, config_dict=config_dict
                )
                dataloader = DataLoader(dataset, batch_size=batch_size,
                                        shuffle=False, num_workers=0)
                with torch.no_grad():
                    for batch_tensor in dataloader:
                        if batch_tensor is None or batch_tensor.size(0) == 0:
                            continue
                        outputs = model(batch_tensor)
                        image_preds.extend(outputs.argmax(dim=1).cpu().numpy())
            if len(image_preds) != len(label_ids):
                raise ValueError("The number of predictions does not match the nuclei count.")
            if save_prediction:
                os.makedirs(predictions_dir, exist_ok=True)
                save_colored_predictions_downsample(
                    labels=mask_data, predictions=image_preds, used_labels=label_ids,
                    myotube_image=image_data, output_path=prediction_out,
                    factor=downsampling_factor
                )

            n_total = len(image_preds)
            n_in = int(np.sum(np.asarray(image_preds) == 1))
            row = {
                "Image Name": stem,
                "Total Number of Nuclei": n_total,
                "Nuclei In": n_in,
                "Nuclei Out": n_total - n_in,
                "Fusion Index (%)": 100 * n_in / n_total if n_total else None,
                "Status": "Complete" if n_total else "No nuclei detected",
            }
            # Save the result before marking the image as classified.
            results[filename] = {"signature": signature, "row": row}
            save_checkpoint(result_store_path, results)
            mark_complete(class_checkpoint_path, class_checkpoint, filename, signature)
            valid_rows.append(row)
        except Exception as error:
            class_checkpoint.get("records", {}).pop(filename, None)
            save_checkpoint(class_checkpoint_path, class_checkpoint)
            outcome="failed"
            print(f"Failed to classify {filename}: {error}")
        finally:
            emit({"kind":"file","filename":filename,"state":outcome,"current":index,"total":len(myotube_files)})

    columns = ["Image Name", "Total Number of Nuclei", "Nuclei In",
               "Nuclei Out", "Fusion Index (%)", "Status"]
    output_path = os.path.join(parent_directory_value, "Fusion_index.xlsx")
    export_path = output_path + ".tmp.xlsx"
    pd.DataFrame(valid_rows, columns=columns).to_excel(export_path, index=False)
    os.replace(export_path, output_path)
    print(f"Results exported to: {output_path} ({len(valid_rows)}/{len(myotube_files)} images)")
    if len(valid_rows) == len(myotube_files):
        display(Markdown("<h3 style='text-align:center'>✅ <b>Classification complete!</b></h3>"))
    else:
        display(Markdown("<h3 style='text-align:center'>⚠️ <b>Partial results: review skipped images.</b></h3>"))


class _MyoFuseProgress:
    def __init__(self):
        self.title=widgets.HTML('Ready')
        self.bar=widgets.IntProgress(value=0,min=0,max=1,layout=widgets.Layout(width='100%',height='17px'))
        self.items=widgets.HTML('<div class="mf-hint">The processed file list will appear here.</div>')
        self.counter=widgets.HTML('')
        self.states={}
        self.box=widgets.VBox([self.title,self.bar,self.counter,self.items])
        self.box.add_class('mf-progress-panel')
    def reset(self):
        self.states.clear();self.title.value='Preparing analysis';self.bar.max=1;self.bar.value=0
        self.items.value='';self.counter.value=''
    def consume(self,event):
        if event['kind']=='stage':
            self.states.clear();self.bar.value=0;self.bar.max=max(1,event['total'])
            self.title.value='<b>'+html.escape(event['stage'].title())+'</b>'
        elif event['kind']=='file':
            self.states[event['filename']]=event['state']
            self.bar.value=min(self.bar.max,event['current'])
            self.counter.value=f"{event['current']} / {event['total']} files"
            self.items.value='<div class="mf-file-list">'+''.join(
               '<div class="mf-file"><span>'+html.escape(name)+'</span><b>'+html.escape(state)+'</b></div>'
               for name,state in reversed(list(self.states.items())[-150:]))+'</div>'

def _preview_crop_rect(width,height,size,xpct,ypct):
    w=min(size,width);h=min(size,height)
    x=max(0,min(width-w,round(width*xpct/100-w/2)))
    y=max(0,min(height-h,round(height*ypct/100-h/2)))
    return int(y),int(y+h),int(x),int(x+w)

def _preview_source(folder,split,myotube_channel,nuclei_channel,size,xpct,ypct):
    if split:
        files=sorted(n for n in os.listdir(folder) if n.lower().endswith(('.czi','.tif','.tiff')))
        if not files:raise ValueError('No original image found.')
        filename=files[0]
        obj=AICSImage(os.path.join(folder,filename),reconstruct_mosaic=False)
        y0,y1,x0,x1=_preview_crop_rect(obj.dims.X,obj.dims.Y,size,xpct,ypct)
        if obj.dims.C>max(myotube_channel,nuclei_channel):
            try:
                lazy=obj.get_image_dask_data('CYX',T=0,Z=0)
                dat=lazy[[myotube_channel,nuclei_channel],y0:y1,x0:x1].compute()
                return filename,np.asarray(dat[0]),np.asarray(dat[1]),(y0,y1,x0,x1)
            except Exception:pass
        tube,nuclei=read_channel_pair(os.path.join(folder,filename),myotube_channel,nuclei_channel)
    else:
        source=os.path.join(folder,'Images')
        files=sorted(n for n in os.listdir(source) if n.lower().endswith(('.tif','.tiff')))
        if not files:raise ValueError('No prepared Images found.')
        filename=files[0]
        tube=imread(os.path.join(source,filename));nuclei=imread(os.path.join(folder,'Nuclei',filename))
    if tube.ndim!=2 or nuclei.shape!=tube.shape:raise ValueError('The selected channels must be matching 2D arrays.')
    y0,y1,x0,x1=_preview_crop_rect(tube.shape[1],tube.shape[0],size,xpct,ypct)
    return filename,tube[y0:y1,x0:x1].copy(),nuclei[y0:y1,x0:x1].copy(),(y0,y1,x0,x1)

def _gray_rgb(plane):
    arr=np.asarray(plane,dtype=np.float32)
    vals=arr[np.isfinite(arr)]
    lo,hi=np.percentile(vals,[1,99.5]) if vals.size else (0.,1.)
    arr=np.uint8(np.clip((arr-lo)/max(1.e-6,hi-lo),0,1)*255)
    return np.repeat(arr[:,:,None],3,axis=2)

def _segmentation_overlay(nuclear_rgb, labels, alpha=0.55):
    """Preview Cellpose labels as reproducibly random colors on nuclear grayscale.

    Background retains its original grayscale intensity. Each nonzero label
    receives one color, filled translucently; white boundaries distinguish
    touching nuclei. This is display-only (no effect on analysis masks).
    """
    labels=np.asarray(labels)
    result=np.asarray(nuclear_rgb,dtype=np.uint8).copy()
    if labels.ndim!=2 or result.shape[:2]!=labels.shape or result.ndim!=3 or result.shape[2]!=3:
        raise ValueError('Segmentation overlay requires matching 2D labels and RGB signal.')
    positive=labels>0
    if not np.any(positive):return result
    unique,inverse=np.unique(labels[positive],return_inverse=True)
    # Seeded RNG keeps colors stable when the same crop is previewed again.
    random=np.random.default_rng(14673)
    hues=random.uniform(0.,1.,size=len(unique))
    palette=np.asarray([colorsys.hsv_to_rgb(float(h),.83,1.) for h in hues],dtype=np.float32)*255.
    background=result[positive].astype(np.float32)
    result[positive]=np.clip((1.-alpha)*background+alpha*palette[inverse],0,255).astype(np.uint8)
    result[find_boundaries(labels,mode='inner')]=(250,250,250)
    return result

def _to_png(rgb):
    image=PILImage.fromarray(rgb.astype(np.uint8),'RGB')
    image.thumbnail((900,900))
    stream=io.BytesIO();image.save(stream,format='PNG');return stream.getvalue()

def preview_myo_fuse(*,root,split,myotube_channel,nuclei_channel,size,xpct,ypct,dia,
                     run_segmentation,run_classification,seg_model_file,class_model_file,
                     seg_device,class_device,percentile_max,half_patch_size,batch_size):
    if not root or not os.path.isdir(root):raise ValueError('Select an image folder.')
    filename,tube,nuclei,rect=_preview_source(root,split,myotube_channel,nuclei_channel,size,xpct,ypct)
    if run_segmentation:
        if not os.path.isfile(seg_model_file or ''):raise ValueError('Select a Cellpose model for preview.')
        model=models.CellposeModel(gpu=seg_device=='GPU' and torch.cuda.is_available(),pretrained_model=seg_model_file)
        mask=np.asarray(model.eval(nuclei,diameter=dia,channels=[0,0],normalize=True)[0],dtype=np.uint32)
    else:
        masks=os.path.join(root,'Masks',os.path.splitext(filename)[0]+'.tif')
        if not os.path.isfile(masks):raise ValueError('Existing mask not found: '+masks)
        y0,y1,x0,x1=rect;mask=imread(masks)[y0:y1,x0:x1]
    ids,centroids=compute_label_centroids(mask)
    preds=[]
    if run_classification and len(ids):
        if not os.path.isfile(class_model_file or ''):raise ValueError('Select a Svetlana model.')
        cfg_path=os.path.join(os.path.dirname(class_model_file),'Config.json')
        with open(cfg_path,'r') as f:cfg=json.load(f)
        dev=torch.device('cuda' if class_device=='GPU' and torch.cuda.is_available() else 'cpu')
        network=torch.load(class_model_file,map_location=dev,weights_only=False)['model'].to(dev)
        network.eval()
        rgb=np.repeat(tube[:,:,None],3,axis=2).astype(np.float32)
        norm=normalize_image(rgb,percentile_max)
        data=PredictionDataset(norm,mask,ids,centroids,half_patch_size,dev,cfg)
        with torch.no_grad():
            for batch in DataLoader(data,batch_size=batch_size,shuffle=False,num_workers=0):
                preds.extend(network(batch).argmax(dim=1).cpu().numpy().tolist())
        if len(preds)!=len(ids):raise ValueError('Classification and mask labels do not agree.')
    orig=_gray_rgb(nuclei);base=_gray_rgb(tube)
    seg=_segmentation_overlay(orig,mask)
    boundary=find_boundaries(mask,mode='inner')
    classified=base.copy()
    if preds:
        for lb,pred in zip(ids,preds): classified[boundary&(mask==lb)]=(40,210,125) if pred==1 else (244,72,99)
    else:classified[boundary]=(20,226,244)
    ins=int(sum(int(x)==1 for x in preds))
    return {'filename':filename,'rect':rect,'total':len(ids),'inside':ins,'outside':len(preds)-ins,
            'fi':100*ins/len(preds) if preds else None,'original':_to_png(orig),
            'segmentation':_to_png(seg),'classification':_to_png(classified)}

#### SET-UP ####

# The widgets keep the original variable names used by the processing callbacks.
def section_card(number, title, description, children):
    heading = widgets.HTML(value=(
        f'<div class="mf-section-heading">'
        f'<span class="mf-number">{number}</span>'
        f'<div><h2>{title}</h2><p>{description}</p></div>'
        f'</div>'
    ))
    wrapper = widgets.VBox([heading, *children], layout=widgets.Layout(width="100%"))
    wrapper.add_class("mf-card")
    return wrapper


def field_label(title, description):
    return widgets.HTML(value=(
        f'<div class="mf-field-label"><strong>{title}</strong>'
        f'<span>{description}</span></div>'
    ))


folder_chooser = FileChooser(str(default_image_directory()), show_only_dirs=True)
nuclei_model = default_model_file("MyoNuclei")
classifier_model = Path.cwd() / "MyoFuse" / "Models" / "Svetlana" / "MyoFuse.pth"
seg_chooser = FileChooser(str(nuclei_model.parent if nuclei_model else Path.cwd()),
                          filename=nuclei_model.name if nuclei_model else "",
                          select_default=bool(nuclei_model))
class_chooser = FileChooser(str(classifier_model.parent if classifier_model.is_file() else Path.cwd()),
                            filename=classifier_model.name if classifier_model.is_file() else "",
                            select_default=classifier_model.is_file())
for chooser in (folder_chooser, seg_chooser, class_chooser):
    chooser.layout = widgets.Layout(width="100%")
    chooser.add_class("mf-chooser")

parent_directory = folder_chooser
seg_directory = seg_chooser
class_directory = class_chooser

folder_field = widgets.VBox([
    field_label("Image folder", "Select the folder containing your original images."),
    folder_chooser
], layout=widgets.Layout(width="100%"))
seg_field = widgets.VBox([
    field_label("Nuclei segmentation model", "Select your Cellpose model file."),
    seg_chooser
], layout=widgets.Layout(width="100%"))
class_field = widgets.VBox([
    field_label("Myonuclei classifier", "Keep Config.json beside the .pth model file."),
    class_chooser
], layout=widgets.Layout(width="100%"))
files_card = section_card("01", "Choose files",
                          "Point MyoFuse to your images and trained models.",
                          [folder_field, seg_field, class_field])

style = {"description_width": "175px"}
full_width = widgets.Layout(width="520px", max_width="100%")
myotube_channel_widget = widgets.IntText(
    value=0, description="Myotube channel", style=style, layout=full_width)
nuclei_channel_widget = widgets.IntText(
    value=1, description="Nuclei channel", style=style, layout=full_width)
split_images_widget = widgets.Checkbox(
    value=True, description="Split original channels", layout=full_width)
channels_card = section_card("02", "Image channels",
                             "Channel numbering starts at 0. Turn off splitting when using prepared Images and Nuclei folders.",
                             [myotube_channel_widget, nuclei_channel_widget,
                              split_images_widget])

run_segmentation_widget = widgets.Checkbox(
    value=True, description="Run nuclei segmentation", layout=full_width)
dia_widget = widgets.BoundedIntText(
    value=24, min=1, max=2048, description="Cellpose diameter (px)", style=style, layout=full_width)
seg_device_widget = widgets.Dropdown(
    options=["GPU", "CPU"], value="GPU", description="Segmentation device",
    style=style, layout=full_width)
segmentation_card = section_card("03", "Nuclei segmentation",
                                 "Use existing masks by turning this stage off.",
                                 [run_segmentation_widget, dia_widget,
                                  seg_device_widget])

run_classification_widget = widgets.Checkbox(
    value=True, description="Run classification and quantify FI", layout=full_width)
class_device_widget = widgets.Dropdown(
    options=["GPU", "CPU"], value="GPU", description="Classifier device",
    style=style, layout=full_width)
percentile_max_widget = widgets.FloatText(
    value=99.7, description="Upper percentile", style=style, layout=full_width)
half_patch_size_widget = widgets.IntText(
    value=100, description="Half patch size", style=style, layout=full_width)
batch_size_widget = widgets.IntText(
    value=512, description="Batch size", style=style, layout=full_width)
classification_card = section_card("04", "Classification",
                                   "The half patch size should match the classifier training setup.",
                                   [run_classification_widget, class_device_widget,
                                    percentile_max_widget, half_patch_size_widget,
                                    batch_size_widget])

save_prediction_widget = widgets.Checkbox(
    value=True, description="Save annotated prediction images", layout=full_width)
save_normalization_widget = widgets.Checkbox(
    value=False, description="Save normalized images", layout=full_width)
downsampling_factor_widget = widgets.IntText(
    value=1, description="Preview downsampling", style=style, layout=full_width)
outputs_card = section_card("05", "Outputs",
                            "Larger downsampling factors make annotated previews smaller.",
                            [save_prediction_widget, save_normalization_widget,
                             downsampling_factor_widget])


progress_panel=_MyoFuseProgress()
preview_size_widget=widgets.Dropdown(options=[256,384,512,768,1024],value=512,description='Crop size',style=style,layout=full_width)
preview_x_widget=widgets.IntSlider(value=50,min=0,max=100,description='Crop X (%)',style=style,layout=full_width)
preview_y_widget=widgets.IntSlider(value=50,min=0,max=100,description='Crop Y (%)',style=style,layout=full_width)
diameter_guide_button=widgets.Button(description='Show diameter guide',icon='circle-o')
preview_test_button=widgets.Button(description='Test segmentation + classification on crop',icon='eye')
guide_output=widgets.Image(format='png',layout=widgets.Layout(width='100%',max_width='450px'))
preview_info=widgets.HTML('Preview is limited to the selected crop; no outputs are written.')
preview_original=widgets.Image(format='png',layout=widgets.Layout(width='100%',max_width='100%'))
preview_segmentation=widgets.Image(format='png',layout=widgets.Layout(width='100%',max_width='100%'))
preview_classification=widgets.Image(format='png',layout=widgets.Layout(width='100%',max_width='100%'))
preview_strip=widgets.HBox([
    widgets.VBox([widgets.HTML('<div class="mf-preview-title">Nuclear signal</div>'),preview_original],
                 layout=widgets.Layout(flex='1 1 280px',width='100%',min_width='0',max_width='100%')),
    widgets.VBox([widgets.HTML('<div class="mf-preview-title">Segmentation</div>'),preview_segmentation],
                 layout=widgets.Layout(flex='1 1 280px',width='100%',min_width='0',max_width='100%')),
    widgets.VBox([widgets.HTML('<div class="mf-preview-title">Classification</div>'),preview_classification],
                 layout=widgets.Layout(flex='1 1 280px',width='100%',min_width='0',max_width='100%')),
],layout=widgets.Layout(width='100%',min_width='0',flex_flow='row wrap',gap='12px'))
preview_strip.add_class('mf-preview-strip')
for tile in preview_strip.children:tile.add_class('mf-preview-tile')
preview_card=section_card('04','Diameter and crop preview','Choose nucleus diameter and test the first image without processing the whole field.',
                          [preview_size_widget,preview_x_widget,preview_y_widget,
                           widgets.HBox([diameter_guide_button,preview_test_button],layout=widgets.Layout(flex_flow='row wrap',gap='12px')),
                           guide_output,preview_info,preview_strip])

def _show_guide(_):
    try:
        if not parent_directory.value:raise ValueError('Choose an image folder.')
        name,tube,nuclei,rect=_preview_source(parent_directory.value,split_images_widget.value,
            myotube_channel_widget.value,nuclei_channel_widget.value,preview_size_widget.value,
            preview_x_widget.value,preview_y_widget.value)
        im=PILImage.fromarray(_gray_rgb(nuclei));x,y=im.size[0]//2,im.size[1]//2;r=dia_widget.value/2
        ImageDraw.Draw(im).ellipse((x-r,y-r,x+r,y+r),outline=(255,194,40),width=3)
        guide_output.value=_to_png(np.asarray(im))
        preview_info.value=f'<div class="mf-note">{html.escape(name)} · guide circle {dia_widget.value} px (native image coordinates). Choose a diameter representative of your nuclei.</div>'
    except Exception as exc:preview_info.value='<div class="mf-note">'+html.escape(str(exc))+'</div>'

diameter_guide_button.on_click(_show_guide)

def _start_preview(_):
    if preview_test_button.disabled:return
    preview_test_button.disabled=True;preview_info.value='<div class="mf-note">Running preview inference...</div>'
    async def worker():
        try:
            loop=asyncio.get_running_loop()
            params=dict(root=parent_directory.value,split=split_images_widget.value,
                myotube_channel=myotube_channel_widget.value,nuclei_channel=nuclei_channel_widget.value,
                size=preview_size_widget.value,xpct=preview_x_widget.value,ypct=preview_y_widget.value,
                dia=dia_widget.value,run_segmentation=run_segmentation_widget.value,
                run_classification=run_classification_widget.value,seg_model_file=seg_directory.value,
                class_model_file=class_directory.value,seg_device=seg_device_widget.value,
                class_device=class_device_widget.value,percentile_max=percentile_max_widget.value,
                half_patch_size=half_patch_size_widget.value,batch_size=batch_size_widget.value)
            result=await loop.run_in_executor(None,lambda:preview_myo_fuse(**params))
            preview_original.value=result['original'];preview_segmentation.value=result['segmentation']
            preview_classification.value=result['classification']
            fi='not calculated' if result['fi'] is None else f"{result['fi']:.1f}%"
            preview_info.value=f'<div class="mf-note"><b>{html.escape(result["filename"])}</b> · {result["total"]} detected nuclei · {result["inside"]} in · {result["outside"]} out · crop FI: {fi}<br><em>Diagnostic crop only; border nuclei can have incomplete context.</em></div>'
        except Exception as exc:
            preview_info.value='<div class="mf-note">Preview failed: '+html.escape(str(exc))+'</div>'
        finally:preview_test_button.disabled=False
    try:asyncio.get_running_loop().create_task(worker())
    except RuntimeError:preview_test_button.disabled=False;preview_info.value='No Jupyter event loop available.'
preview_test_button.on_click(_start_preview)

run_button = widgets.Button(description="Launch new analysis", icon="play",
                            button_style="success",
                            layout=widgets.Layout(width="225px", height="46px"))
resume_button = widgets.Button(description="Resume analysis", icon="repeat",
                               layout=widgets.Layout(width="205px", height="46px"))
run_button.add_class("mf-primary-button")
resume_button.add_class("mf-secondary-button")
buttons_box = widgets.HBox([run_button, resume_button],
                           layout=widgets.Layout(width="100%", flex_flow="row wrap"))
buttons_box.add_class("mf-buttons")
status_badge = widgets.HTML(
    value='<div class="mf-status mf-ready"><span class="mf-status-dot"></span>Ready to analyze</div>'
)
output = widgets.Output(layout=widgets.Layout(width="100%"))
output.add_class("mf-output")
run_card = section_card("06", "Run analysis",
                        "Launch begins a fresh run and backs up previous results. Resume checks saved results and processes what is missing or changed.",
                        [buttons_box, status_badge, progress_panel.box, output])

header = widgets.HTML(value='''
<div class="mf-hero">
  <div class="mf-hero-top"><span class="mf-mark">MF</span><span class="mf-version">VERSION 1.1.0</span></div>
  <h1>MyoFuse</h1>
  <p>Nuclei segmentation, myonuclei classification and fusion index quantification in one workflow.</p>
  <div class="mf-hero-foot"><span>01 &nbsp;Choose files</span><span>02 &nbsp;Set parameters</span><span>03 &nbsp;Analyze</span><a class="mf-paper-link" href="https://doi.org/10.1038/s41598-026-40047-y" target="_blank" rel="noopener noreferrer">Read the publication ↗</a></div>
</div>
''')

main_ui = widgets.VBox(
    [header, files_card, channels_card, segmentation_card,
     preview_card, classification_card, outputs_card, run_card],
    layout=widgets.Layout(width="100%")
)
main_ui.add_class("myofuse-app")


def _set_status(message, state):
    status_badge.value = (
        f'<div class="mf-status mf-{state}">'
        f'<span class="mf-status-dot"></span>{message}</div>'
    )


def _widget_values():
    return (
        myotube_channel_widget.value,
        nuclei_channel_widget.value,
        dia_widget.value,
        percentile_max_widget.value,
        half_patch_size_widget.value,
        batch_size_widget.value,
        split_images_widget.value,
        run_segmentation_widget.value,
        seg_device_widget.value,
        run_classification_widget.value,
        class_device_widget.value,
        save_normalization_widget.value,
        save_prediction_widget.value,
        downsampling_factor_widget.value,
        parent_directory.value,
        seg_directory.value,
        class_directory.value,
    )


def _start_analysis(resume=False):
    if run_button.disabled:return
    run_button.disabled=True;resume_button.disabled=True
    progress_panel.reset();_set_status('Analysis running','running')
    params=_widget_values()
    async def worker():
        try:
            loop=asyncio.get_running_loop()
            def go():
                if not resume:reset_checkpoints(parent_directory.value)
                return run_entire_analysis(*params,emit=lambda event:loop.call_soon_threadsafe(progress_panel.consume,event))
            await loop.run_in_executor(None,go)
            _set_status('Run finished; inspect file statuses and results','complete')
        except Exception as exc:
            _set_status('Run failed; inspect error below','error')
            with output:print('Error:',exc);traceback.print_exc()
        finally:run_button.disabled=False;resume_button.disabled=False
    try:asyncio.get_running_loop().create_task(worker())
    except RuntimeError:
        run_button.disabled=False;resume_button.disabled=False
        _set_status('Unable to start Jupyter event loop','error')


run_button.on_click(lambda button: _start_analysis(resume=False))
resume_button.on_click(lambda button: _start_analysis(resume=True))
display(main_ui)
```
