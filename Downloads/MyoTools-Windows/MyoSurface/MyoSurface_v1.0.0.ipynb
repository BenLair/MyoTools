```python
"""MyoSurface computational functions, embedded in the Voilà notebook on export.

Scientific correspondence: Fiji myotube_surface_v3_260828.groovy.
The Fiji rolling ball uses downsampled minimum blocks, a trimmed spherical patch,
non-flat grayscale opening, then linear interpolation. This Python approximation
replaces the Java loop with SciPy morphology, but still requires paired-image validation.
"""

from __future__ import annotations

import hashlib
import json
import os
import shutil
import sys
from dataclasses import asdict, dataclass
from datetime import datetime, timezone
from importlib.metadata import PackageNotFoundError, version as dist_version
from pathlib import Path

import numpy as np
import pandas as pd
import scipy
from scipy import ndimage as ndi
import skimage
import tifffile
from tifffile import imwrite

APP_VERSION = "1.0.0"
SUPPORTED_EXTENSIONS = {".tif", ".tiff", ".czi"}
PIPELINE_ID = "myosurface-fiji-inspired-v2-acquired-roi"


@dataclass(frozen=True)
class SurfaceParameters:
    channel: int = 1                 # User-facing Fiji 1-based channel index
    scene: int = 0                   # Zero-based AICS scene index
    timepoint: int = 0               # Zero-based T
    z_plane: int = 0                 # Zero-based Z
    reconstruct_mosaic: bool = True
    background_radius: int = 200
    background_mean: float = 3500.0
    threshold: float = 200.0
    min_area: int = 500
    median_radius: int = 1
    connectivity: int = 1            # MorphoLibJ default: 4-connected (1); optional 8 (2)
    save_masks: bool = True
    save_corrected: bool = False

    def validate(self):
        if self.channel < 1:
            raise ValueError("Myotube channel must be >= 1 (Fiji-style numbering).")
        for name in ("scene", "timepoint", "z_plane"):
            if getattr(self, name) < 0:
                raise ValueError(f"{name} must be >= 0.")
        if not (1 <= self.background_radius <= 2000):
            raise ValueError("Rolling-ball radius must be between 1 and 2000 pixels.")
        if not np.isfinite(self.background_mean) or self.background_mean < 0:
            raise ValueError("Background mean must be a finite, nonnegative number.")
        if not np.isfinite(self.threshold) or self.threshold < 0:
            raise ValueError("Threshold must be a finite, nonnegative number.")
        if self.min_area < 1 or not (0 <= self.median_radius <= 25):
            raise ValueError("Minimum area must be >= 1 and median radius 0..25.")
        if self.connectivity not in (1, 2):
            raise ValueError("Connectivity must be 1 (4-connected) or 2 (8-connected).")


def installed_versions():
    packages = (
        "numpy", "scipy", "scikit-image", "aicsimageio", "aicspylibczi",
        "tifffile", "pandas", "ipywidgets", "ipyfilechooser", "voila",
    )
    result = {"python": sys.version.split()[0], "myosurface": APP_VERSION}
    for package in packages:
        try:
            result[package] = dist_version(package)
        except PackageNotFoundError:
            result[package] = "not installed"
    return result


def list_images(folder):
    source = Path(folder).expanduser()
    if not source.is_dir():
        raise ValueError("Choose an existing image folder.")
    return sorted((p for p in source.iterdir()
                   if p.is_file() and p.suffix.lower() in SUPPORTED_EXTENSIONS),
                  key=lambda p: p.name.casefold())


def _extract_tiff_plane(path, scene, timepoint, z_plane, channel):
    """Explicit-axis TIFF fallback. Reject ambiguous stacks rather than silently misread."""
    with tifffile.TiffFile(path) as tf:
        series_count = len(tf.series)
        if scene >= series_count:
            raise ValueError(f"TIFF scene {scene} unavailable ({len(tf.series)} series).")
        series = tf.series[scene]
        axes = series.axes.upper()
        data = series.asarray()

    # The channel dimension in ordinary greyscale data is C. RGB TIFFs use S.
    relevant = [key for key in "CS" if key in axes and data.shape[axes.index(key)] > 1]
    if len(relevant) > 1:
        raise ValueError("This TIFF has both C and S dimensions; channel selection is ambiguous.")
    channel_axis = relevant[0] if relevant else None
    requested = {"T": timepoint, "Z": z_plane, "C": channel - 1, "S": channel - 1}
    selection = []
    for axis, size in zip(axes, data.shape):
        if axis in "YX":
            selection.append(slice(None))
        elif axis == channel_axis:
            index = requested[axis]
            if index >= size:
                raise ValueError(f"Requested channel {channel}; available channels: {size}.")
            selection.append(index)
        elif axis in "TZ":
            index = requested[axis]
            if index >= size:
                raise ValueError(f"Requested {axis}={index}; available {axis}: {size}.")
            selection.append(index)
        elif size == 1:
            selection.append(0)
        else:
            raise ValueError(f"Unknown, non-singleton TIFF axis {axis!r}; refusing to guess.")
    if channel_axis is None and channel != 1:
        raise ValueError("Only one channel is present in this TIFF.")
    if "T" not in axes and timepoint:
        raise ValueError("TIFF has no time axis (T=0 required).")
    if "Z" not in axes and z_plane:
        raise ValueError("TIFF has no Z axis (Z=0 required).")
    plane = np.asarray(data[tuple(selection)])
    remainder = "".join(ax for ax, sel in zip(axes, selection) if isinstance(sel, slice))
    if set(remainder) != {"X", "Y"} or plane.ndim != 2:
        raise ValueError(f"Cannot interpret TIFF dimensions {axes!r} as one XY plane.")
    if remainder == "XY":
        plane = plane.T
    metadata = {"reader": "tifffile (fallback)", "scene_count": series_count,
                "original_axes": axes, "original_shape": list(data.shape)}
    return plane, metadata


def load_channel(path, params):
    """Load one full-resolution YX plane. No tile-by-tile segmentation."""
    params.validate()
    path = Path(path)
    if not path.is_file() or path.suffix.lower() not in SUPPORTED_EXTENSIONS:
        raise ValueError(f"Invalid or unsupported image: {path}")
    try:
        from aicsimageio import AICSImage
    except ImportError as error:
        if path.suffix.lower() == ".czi":
            raise ImportError("CZI reading requires aicsimageio and aicspylibczi.") from error
        image, metadata = _extract_tiff_plane(path, params.scene, params.timepoint,
                                              params.z_plane, params.channel)
    else:
        try:
            reader = AICSImage(str(path), reconstruct_mosaic=params.reconstruct_mosaic)
        except Exception as error:
            if path.suffix.lower() != ".czi":
                image, metadata = _extract_tiff_plane(path, params.scene,
                    params.timepoint, params.z_plane, params.channel)
            else:
                raise RuntimeError(
                    f"CZI reader failed for {path.name}. Check the CZI plugin, "
                    "scene selection and mosaic reconstruction."
                ) from error
        else:
            if params.scene >= len(reader.scenes):
                raise ValueError(f"Scene {params.scene} unavailable; found {len(reader.scenes)} scene(s).")
            reader.set_scene(params.scene)
            sizes = reader.dims
            if "M" in sizes.order and getattr(sizes, "M", 1) > 1:
                raise ValueError(
                    "This file contains separate mosaic tiles (M dimension). "
                    "Enable mosaic reconstruction to analyze the complete image; "
                    "MyoSurface will not silently measure only one tile."
                )
            c_count = int(getattr(sizes, "C", 1))
            s_count = int(getattr(sizes, "S", 1))
            if c_count > 1 and s_count > 1:
                raise ValueError("Both C and S axes have multiple channels; please export "
                                 "one fluorescence-channel image per C axis.")
            channel_axis = "C" if c_count > 1 or s_count == 1 else "S"
            max_channels = c_count if channel_axis == "C" else s_count
            if params.channel > max_channels:
                raise ValueError(f"Channel {params.channel} unavailable; found "
                                 f"{max_channels} channels on axis {channel_axis}.")
            for dim, requested in (("T", params.timepoint), ("Z", params.z_plane)):
                maximum = int(getattr(sizes, dim, 1))
                if requested >= maximum:
                    raise ValueError(f"{dim} index {requested} unavailable; size={maximum}.")
            # TIFF RGB images often use a sample axis S rather than C.
            selectors = {"C": params.channel-1 if channel_axis == "C" else 0,
                         "T": params.timepoint, "Z": params.z_plane}
            if "S" in sizes.order:
                selectors["S"] = params.channel-1 if channel_axis == "S" else 0
            if "M" in sizes.order:
                selectors["M"] = 0
            try:
                # Dask selects the requested plane BEFORE loading it into RAM.
                # This is NOT tile-wise segmentation; the complete selected YX plane
                # is processed as one array at its native resolution.
                image = np.asarray(reader.get_image_dask_data(
                    "YX", **selectors).compute())
            except Exception as error:
                raise RuntimeError(
                    f"Failed to read a single YX plane from {path.name}. "
                    "Try another scene or disable mosaic reconstruction if not appropriate."
                ) from error
            metadata = {"reader": "AICSImageIO", "scene_count": len(reader.scenes),
                        "scene_name": str(reader.current_scene),
                        "dims": {k: int(getattr(sizes, k, 1)) for k in ("X", "Y", "C", "T", "Z", "S")},
                        "channel_axis": channel_axis}
    if image.ndim != 2:
        raise ValueError(f"Expected a single 2D XY plane; received shape {image.shape}.")
    if image.dtype.kind not in "uif":
        raise TypeError(f"Unsupported image dtype {image.dtype}; expected 8/16-bit or float grayscale.")
    if image.dtype.kind == "f" and not np.isfinite(image).all():
        raise ValueError("The image contains NaN or infinite pixel values.")
    if np.any(image < 0):
        raise ValueError("Negative intensities are not supported by this fluorescence pipeline.")
    metadata["shape_yx"] = [int(v) for v in image.shape]
    metadata["dtype"] = str(image.dtype)
    return image, metadata


def _fiji_ball_settings(radius):
    if radius <= 10:
        return 1, 24
    if radius <= 30:
        return 2, 24
    if radius <= 100:
        return 4, 32
    return 8, 40


def _fiji_ball_structure(radius, shrink, trim_percent):
    small_r = max(1.0, radius / shrink)
    trim = int(trim_percent * small_r / 100)
    # Java Math.round(x) for positive x: floor(x + 0.5), not Python round.
    halfwidth = int(np.floor((small_r - trim) + 0.5))
    coords = np.arange(-halfwidth, halfwidth + 1, dtype=np.float32)
    xx, yy = np.meshgrid(coords, coords)
    patch = np.sqrt(np.clip(small_r**2 - xx**2 - yy**2, 0, None))
    return patch.astype(np.float32)


def _block_min(array, factor):
    if factor == 1:
        return np.asarray(array, dtype=np.float32)
    out = np.minimum.reduceat(array, np.arange(0, array.shape[0], factor), axis=0)
    out = np.minimum.reduceat(out, np.arange(0, array.shape[1], factor), axis=1)
    return np.asarray(out, dtype=np.float32)


def _fiji_bilinear_expand(image, shape, factor, block_rows=128):
    """Pixel-centre bilinear upsampling with linear border extrapolation (Fiji-like)."""
    h, w = shape
    small_h, small_w = image.shape
    if small_h < 2 or small_w < 2:
        # Degenerate images have no two adjacent sample points for extrapolation.
        return ndi.zoom(image, (h / small_h, w / small_w),
                        order=1, mode="nearest", grid_mode=True).astype(np.float32)
    xs = (np.arange(w, dtype=np.float32) + 0.5) / factor - 0.5
    ys = (np.arange(h, dtype=np.float32) + 0.5) / factor - 0.5
    x0 = np.clip(np.floor(xs).astype(np.intp), 0, small_w - 2)
    y0 = np.clip(np.floor(ys).astype(np.intp), 0, small_h - 2)
    wx = (xs - x0)[None, :]
    output = np.empty(shape, dtype=np.float32)
    for start in range(0, h, block_rows):
        end = min(h, start + block_rows)
        rows = y0[start:end, None]
        wy = (ys[start:end, None] - rows).astype(np.float32)
        # Separable, vectorized bilinear interpolation with extrapolation near edges.
        top = (1 - wx) * image[rows, x0[None, :]] + wx * image[rows, x0[None, :] + 1]
        bottom = ((1 - wx) * image[rows + 1, x0[None, :]]
                  + wx * image[rows + 1, x0[None, :] + 1])
        output[start:end] = (1 - wy) * top + wy * bottom
    return output


def estimate_fiji_style_background(image, radius):
    """Fiji-inspired rolling ball, accelerated with SciPy nonflat gray morphology.

    Fiji ImageJ 1.x: separable 3x3 mean filter; block-min decimation with
    radius-dependent factor; rolling a trimmed spherical patch under the
    decimated image; bilinear upsampling. Not a claim of bit-for-bit identity.
    """
    if image.ndim != 2:
        raise ValueError("Background subtraction supports single-channel 2D arrays only.")
    if radius < 1:
        raise ValueError("Background radius must be >= 1.")
    smooth = ndi.uniform_filter(np.asarray(image, dtype=np.float32),
                                size=3, mode="nearest")
    shrink, trim = _fiji_ball_settings(radius)
    small = _block_min(smooth, shrink)
    ball = _fiji_ball_structure(radius, shrink, trim)
    # Fiji lets the ball centre run outside the image extent. Padding permits
    # these centres while suppressing out-of-image pixels during erosion.
    halfwidth = ball.shape[0] // 2
    padded = np.pad(small, halfwidth, mode="constant", constant_values=np.inf)
    erosion = ndi.grey_erosion(padded, structure=ball, mode="constant", cval=np.inf)
    opening = ndi.grey_dilation(erosion, structure=ball,
                                mode="constant", cval=-np.inf)
    small_bg = opening[halfwidth:halfwidth + small.shape[0],
                       halfwidth:halfwidth + small.shape[1]]
    if shrink == 1:
        return small_bg.astype(np.float32, copy=False)
    return _fiji_bilinear_expand(small_bg, image.shape, shrink)


def _fiji_median_footprint(radius):
    """ImageJ RankFilters circular footprint; radius=1 includes the full 3x3."""
    r2 = radius * radius + 1
    kr = int(np.sqrt(r2 + 1e-10))
    footprint = np.zeros((2 * kr + 1, 2 * kr + 1), dtype=bool)
    for y in range(-kr, kr + 1):
        dx = int(np.sqrt(r2 - y * y + 1e-10))
        footprint[y + kr, kr - dx:kr + dx + 1] = True
    return footprint


def _remove_small_binary_components(binary, area, connectivity):
    """Keep components whose pixel count is >= area (MorphoLibJ-style)."""
    if area <= 1:
        return np.asarray(binary, dtype=bool)
    structure = ndi.generate_binary_structure(2, connectivity)
    component_labels, count = ndi.label(np.asarray(binary, dtype=bool), structure)
    if count == 0:
        return np.zeros_like(binary, dtype=bool)
    sizes = np.bincount(component_labels.ravel())
    keep = sizes >= area
    keep[0] = False
    return keep[component_labels]


def estimate_zero_replacement(original, target_samples=1000000):
    """Estimate the nonzero background level with a bounded, robust sample.

    The mean of the 10th-50th percentile band of NONZERO source intensities
    reduces contamination from bright myotubes. It is a suggested baseline,
    not a mathematically unique or scientifically validated ground truth.
    This function is called only on user request; the suggested value is
    copied into one fixed parameter shared by the whole batch.
    """
    image = np.asarray(original)
    if image.ndim != 2 or image.dtype.kind not in "uif":
        raise ValueError("Expected a two-dimensional nonnegative fluorescence image.")
    if np.any(image < 0) or not np.isfinite(image).all():
        raise ValueError("Input contains negative or non-finite pixel values.")
    stride = max(1, int(np.ceil(np.sqrt(image.size / float(target_samples)))))
    sampled = image[::stride, ::stride]
    nonzero = sampled[sampled > 0]
    if nonzero.size < 8:
        raise ValueError("Too few nonzero sample pixels to estimate the background; choose manual replacement.")
    p10, p50, p90 = np.percentile(nonzero, [10, 50, 90])
    lower_background = nonzero[(nonzero >= p10) & (nonzero <= p50)]
    estimate = float(np.mean(lower_background, dtype=np.float64))
    if image.dtype.kind in "ui":
        estimate = float(np.floor(np.clip(estimate, 0, np.iinfo(image.dtype).max)))
    return {
        "value": estimate,
        "method": "mean of 10th–50th percentile of nonzero intensity samples",
        "positive_samples": int(nonzero.size),
        "stride": int(stride),
        "p10": float(p10), "p50": float(p50), "p90": float(p90),
        "nonzero_fraction_sample": float(nonzero.size / sampled.size),
    }


def resolve_zero_replacement(original, params):
    """Read the same user-selected batch value for every image (Fiji dtype semantics)."""
    value = float(params.background_mean)
    if original.dtype.kind in "ui":
        max_val = float(np.iinfo(original.dtype).max)
        if value > max_val:
            raise ValueError(
                f"Batch zero-pixel replacement {value:g} exceeds this image type's "
                f"maximum ({max_val:g} for {original.dtype}). Choose a common "
                "value appropriate for every image instead of silently clipping it."
            )
        value = float(np.floor(value))
    return value


def prepare_surface_corrected(original, params, replacement=None):
    """Perform the SAME background correction for full runs and crop previews."""
    original = np.asarray(original)
    if replacement is None:
        replacement = resolve_zero_replacement(original, params)
    filled = original.astype(np.float32, copy=True)
    filled[original == 0] = float(replacement)
    bg = estimate_fiji_style_background(filled, params.background_radius)
    corrected = np.clip(filled - bg, 0, None)
    if original.dtype.kind in "ui":
        max_val = float(np.iinfo(original.dtype).max)
        corrected = np.clip(np.floor(corrected + 0.5), 0, max_val).astype(original.dtype)
    else:
        corrected = corrected.astype(np.float32)
    return corrected


def segment_surface_corrected(corrected, original_valid, params, threshold=None):
    """Fiji-inspired median and area filtering, restricted to truly acquired pixels.

    The acquisition ROI is computed from the RAW image before zero replacement.
    Provisional positives on original zeros are always excluded from the final
    mask; there is deliberately no selectable denominator mode.
    """
    threshold = params.threshold if threshold is None else float(threshold)
    if not np.isfinite(threshold) or threshold < 0:
        raise ValueError("Threshold must be finite and nonnegative.")
    acquired = np.asarray(original_valid, dtype=bool)
    if acquired.shape != corrected.shape:
        raise ValueError("Original acquisition ROI and corrected image have different shapes.")
    binary = (corrected >= threshold) & (corrected <= 70000)
    if params.median_radius > 0:
        binary = ndi.median_filter(binary, footprint=_fiji_median_footprint(params.median_radius),
                                   mode="nearest")
    binary = _remove_small_binary_components(binary, params.min_area, params.connectivity)
    binary = ~_remove_small_binary_components(~binary, params.min_area, params.connectivity)
    # The filled zeros were only for rolling-ball background estimation.
    # They NEVER contribute to the final numerator.
    return binary & acquired


def prepare_threshold_crop(image, params, crop_size=768, x_percent=50, y_percent=50):
    """Compute corrected pixels once on a halo-padded crop; never tile the batch.

    A rolling-ball halo and shrink-grid alignment reduce crop-boundary artifacts.
    Cropped estimates are still a QC approximation, particularly near mosaic
    boundaries or when myotubes connect beyond the crop. Final batch calculations
    always use the full original image.
    """
    image = np.asarray(image)
    if image.ndim != 2:
        raise ValueError("Expected a single 2D image.")
    height, width = image.shape
    side = int(crop_size)
    if side < 64:
        raise ValueError("Preview crop must be at least 64 pixels wide.")
    x_percent, y_percent = float(x_percent), float(y_percent)
    if not (0 <= x_percent <= 100 and 0 <= y_percent <= 100):
        raise ValueError("Crop centre percentages must be between 0 and 100.")
    halfw, halfh = min(width, side), min(height, side)
    cx = int(round((width - 1) * x_percent / 100))
    cy = int(round((height - 1) * y_percent / 100))
    x0 = max(0, min(width - halfw, cx - halfw // 2))
    y0 = max(0, min(height - halfh, cy - halfh // 2))
    x1, y1 = x0 + halfw, y0 + halfh
    # Use a halo around the crop for the rolling-ball background fit.
    shrink, _ = _fiji_ball_settings(params.background_radius)
    halo = 2 * int(params.background_radius) + 2 * shrink + 4
    sx0 = max(0, ((x0 - halo) // shrink) * shrink)
    sy0 = max(0, ((y0 - halo) // shrink) * shrink)
    sx1 = min(width, x1 + halo)
    sy1 = min(height, y1 + halo)
    replacement = resolve_zero_replacement(image, params)
    prepared = prepare_surface_corrected(image[sy0:sy1, sx0:sx1], params, replacement)
    cropped = np.array(prepared[y0-sy0:y1-sy0, x0-sx0:x1-sx0], copy=True)
    original_crop = np.array(image[y0:y1, x0:x1], copy=True)
    return {
        "corrected": cropped,
        "original": original_crop,
        "source_patch": np.array(image[sy0:sy1, sx0:sx1], copy=True),
        "source_dtype": str(image.dtype),
        "crop_in_patch": (x0-sx0, y0-sy0, x1-sx0, y1-sy0),
        "replacement": float(replacement),
        "crop_bounds": (x0, y0, x1, y1),
        "halo_px": int(halo),
        "source_shape": image.shape,
    }


def refresh_threshold_crop_replacement(item, params):
    """Recompute preview background correction after changing one batch-wide value.

    Reuses the raw halo-padded crop saved by prepare_threshold_crop, therefore
    a manual edit does not reload the full source image. The production analysis
    still calculates its rolling-ball background on every *whole* image.
    """
    params.validate()
    source = item["source_patch"]
    replacement = resolve_zero_replacement(source, params)
    corrected = prepare_surface_corrected(source, params, replacement)
    x0, y0, x1, y1 = item["crop_in_patch"]
    updated = dict(item)
    updated["corrected"] = np.array(corrected[y0:y1, x0:x1], copy=True)
    updated["replacement"] = float(replacement)
    return updated


def quantify_surface(original, params):
    """Myotube coverage of the acquired area, calculated at native resolution.

    Coverage (%) = 100 * count(segmented AND original > 0) / count(original > 0).
    An original zero is treated as an unacquired mosaic pixel and is excluded
    from BOTH numerator and denominator, even when it was temporarily replaced
    for rolling-ball background correction. All batch parameters remain fixed.
    """
    params.validate()
    original = np.asarray(original)
    if original.ndim != 2:
        raise ValueError("Expected a 2D single-channel image.")
    if original.dtype.kind not in "uif":
        raise TypeError("Only unsigned integer or float fluorescence images are supported.")
    if np.any(original < 0) or not np.isfinite(original).all():
        raise ValueError("Input contains negative or non-finite pixel values.")

    acquired = original > 0
    total_count = int(original.size)
    acquired_count = int(np.count_nonzero(acquired))
    if acquired_count == 0:
        raise ValueError("No acquired pixels (original intensity > 0): surface coverage undefined.")

    replacement = resolve_zero_replacement(original, params)
    corrected = prepare_surface_corrected(original, params, replacement)
    binary = segment_surface_corrected(corrected, acquired, params)
    foreground_count = int(np.count_nonzero(binary))
    assert foreground_count <= acquired_count  # physical area coverage cannot exceed 100%

    return {
        "mask": binary,
        "corrected": corrected,
        "area_percent": 100.0 * foreground_count / acquired_count,
        "foreground_pixels": foreground_count,
        "valid_pixels": acquired_count,
        "total_pixels": total_count,
        "excluded_pixels": total_count - acquired_count,
        "zero_pixel_replacement_used": float(replacement),
    }


def _atomic_json(path, obj):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    temporary = path.with_name(path.name + ".tmp")
    with open(temporary, "w", encoding="utf-8") as out:
        json.dump(obj, out, indent=2, ensure_ascii=False, allow_nan=False)
    os.replace(temporary, path)


def _read_checkpoint(path):
    path = Path(path)
    if not path.is_file():
        return {"pipeline_id": PIPELINE_ID, "records": {}}
    with open(path, "r", encoding="utf-8") as f:
        data = json.load(f)
    if not isinstance(data.get("records"), dict):
        raise ValueError("Malformed MyoSurface checkpoint: expected a records mapping.")
    if data.get("pipeline_id") != PIPELINE_ID:
        # Previous versions used different coverage denominators and can never
        # be reused as results of the current acquired-ROI method.
        return {"pipeline_id": PIPELINE_ID, "records": {}, "previous_pipeline": data.get("pipeline_id")}
    return data


def _input_signature(path, params):
    path = Path(path)
    stat = path.stat()
    content = {"input": str(path.resolve()), "size": stat.st_size,
               "mtime_ns": stat.st_mtime_ns, "parameters": asdict(params),
               "pipeline": PIPELINE_ID, "versions": {"scipy": scipy.__version__,
               "scikit-image": skimage.__version__}}
    return hashlib.sha256(json.dumps(content, sort_keys=True).encode()).hexdigest()


def _atomic_tiff(path, array):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    temporary = path.with_name(path.name + ".tmp.tif")
    try:
        imwrite(str(temporary), array)
        os.replace(temporary, path)
    finally:
        if temporary.exists():
            temporary.unlink()


def _atomic_csv(path, rows):
    columns = [
        "Filename", "Area percent (%)", "Foreground pixels", "Valid pixels",
        "Total pixels", "Excluded zero pixels", "Height (px)", "Width (px)",
        "Scene (0-based)", "Channel (1-based)", "T (0-based)", "Z (0-based)",
        "Background radius (px)", "Background mean",
        "Zero replacement used", "Threshold", "Minimum area (px2)", "Status", "Error",
    ]
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    temporary = path.with_suffix(".tmp.csv")
    pd.DataFrame(rows, columns=columns).to_csv(temporary, index=False)
    os.replace(temporary, path)


def run_batch(folder, params, resume, progress_callback=None, log_callback=None, file_callback=None):
    """Run or resume a folder analysis. Writes results incrementally and atomically."""
    params.validate()
    folder = Path(folder).expanduser().resolve()
    images = list_images(folder)
    if not images:
        raise ValueError("No supported .tif, .tiff or .czi files in selected folder.")
    output_dir = folder / "MyoSurface_Output"
    masks_dir = output_dir / "Masks"
    corrected_dir = output_dir / "Corrected"
    log_dir = output_dir / "Log"
    log_dir.mkdir(parents=True, exist_ok=True)
    csv_path = output_dir / "MyoSurface_results.csv"
    checkpoint_path = log_dir / "checkpoint.json"

    if not resume:
        if csv_path.exists():
            now = datetime.now().strftime("%Y%m%d_%H%M%S_%f")
            shutil.copy2(csv_path, log_dir / f"results_before_{now}.csv")
        checkpoint = {"pipeline_id": PIPELINE_ID, "records": {}}
        _atomic_json(checkpoint_path, checkpoint)
    else:
        checkpoint = _read_checkpoint(checkpoint_path)
        if checkpoint.get("previous_pipeline") is not None and csv_path.is_file():
            # Preserve previous method's CSV when switching from legacy modes.
            stamp = datetime.now().strftime("%Y%m%d_%H%M%S_%f")
            shutil.copy2(csv_path, log_dir / f"results_before_roi_update_{stamp}.csv")
    manifest = {
        "application": "MyoSurface", "version": APP_VERSION,
        "created_at": datetime.now(timezone.utc).astimezone().isoformat(),
        "input_folder": str(folder), "params": asdict(params),
        "software_versions": installed_versions(),
        "method": "Fiji-inspired downsampled spherical grayscale opening; validation pending",
        "fiji_reference": "myotube_surface_v3_260828.groovy",
        "notes": "Scene, Z, and T are single selected planes; all pixels processed at native resolution.",
        "coverage_formula": "100 * pixels(segmented AND original > 0) / pixels(original > 0)",
        "acquisition_roi": "Pixels exactly equal to zero in the original image are excluded from numerator and denominator",
        "zero_replacement_scope": "Used only before rolling-ball correction, never for acquisition ROI",
    }
    _atomic_json(log_dir / "last_analysis_manifest.json", manifest)
    records = checkpoint["records"]
    total = len(images)
    complete, skipped, failed = 0, 0, 0
    rows = []

    for index, image_path in enumerate(images, start=1):
        filename = image_path.name
        mask_path = masks_dir / (filename + "_mask.tif")
        corrected_path = corrected_dir / (filename + "_corrected.tif")
        parameters_path = log_dir / (filename + "_parameters.json")
        required = [parameters_path]
        if params.save_masks:
            required.append(mask_path)
        if params.save_corrected:
            required.append(corrected_path)
        signature = _input_signature(image_path, params)
        previous = records.get(filename)
        if (resume and previous and previous.get("signature") == signature
                and all(p.is_file() for p in required)):
            row = previous["row"]
            rows.append(row)
            skipped += 1
            if file_callback:
                file_callback(filename, "skipped", row.get("Area percent (%)"), None)
            if log_callback:
                log_callback(f"SKIP   {filename} (unchanged)")
        else:
            if file_callback:
                file_callback(filename, "processing", None, None)
            try:
                image, metadata = load_channel(image_path, params)
                result = quantify_surface(image, params)
                if params.save_masks:
                    _atomic_tiff(mask_path, result["mask"].astype(np.uint8) * 255)
                if params.save_corrected:
                    _atomic_tiff(corrected_path, result["corrected"])
                row = {
                    "Filename": filename,
                    "Area percent (%)": result["area_percent"],
                    "Foreground pixels": result["foreground_pixels"],
                    "Valid pixels": result["valid_pixels"],
                    "Total pixels": result["total_pixels"],
                    "Excluded zero pixels": result["excluded_pixels"],
                    "Height (px)": int(image.shape[0]), "Width (px)": int(image.shape[1]),
                    "Scene (0-based)": params.scene,
                    "Channel (1-based)": params.channel,
                    "T (0-based)": params.timepoint,
                    "Z (0-based)": params.z_plane,
                    "Background radius (px)": params.background_radius,
                    "Background mean": params.background_mean,
                    "Zero replacement used": result["zero_pixel_replacement_used"],
                    "Threshold": params.threshold,
                    "Minimum area (px2)": params.min_area,
                    "Status": "Complete", "Error": "",
                }
                _atomic_json(parameters_path, {
                    **manifest, "file": filename, "input_metadata": metadata,
                    "input_signature": signature,
                    "measurement": {key: value for key, value in result.items()
                                    if key not in ("mask", "corrected")},
                })
                records[filename] = {"signature": signature, "row": row}
                _atomic_json(checkpoint_path, checkpoint)
                complete += 1
                if file_callback:
                    file_callback(filename, "completed", result["area_percent"], None)
                if log_callback:
                    log_callback(f"OK     {filename} → {result['area_percent']:.3f}%")
            except Exception as error:
                records.pop(filename, None)
                _atomic_json(checkpoint_path, checkpoint)
                failed += 1
                row = {"Filename": filename, "Status": "Error", "Error": str(error)}
                if file_callback:
                    file_callback(filename, "failed", None, str(error))
                if log_callback:
                    log_callback(f"ERROR  {filename} → {error}")
            rows.append(row)
        # Always reconstruct a CSV with one row per current input file (no duplicates).
        _atomic_csv(csv_path, rows)
        if progress_callback:
            progress_callback(index, total)

    return {"total": total, "processed": complete, "skipped": skipped,
            "failed": failed, "results_csv": str(csv_path), "output_folder": str(output_dir)}
```

```python
"""MyoSurface ipywidgets interface (Voilà compatible)."""

import asyncio
import html
import io

import ipywidgets as widgets
from ipyfilechooser import FileChooser

def default_image_directory():
    """Use the user's real Desktop, or their home if no Desktop is available.

    This standard-library-only function is embedded in the notebooks so they
    also work when launched directly, without importing or starting the hub.
    """
    import os
    import subprocess
    from pathlib import Path

    if os.name == "nt":
        command = ["powershell.exe", "-NoProfile", "-Command",
                   "[Console]::OutputEncoding = [System.Text.Encoding]::UTF8; [Environment]::GetFolderPath('DesktopDirectory')"]
    else:
        command = ["xdg-user-dir", "DESKTOP"]
    try:
        result = subprocess.run(command, capture_output=True, text=True, timeout=5,
                                encoding="utf-8", errors="replace")
        if result.returncode == 0 and result.stdout.strip():
            desktop = Path(result.stdout.strip().lstrip("\ufeff")).expanduser()
            if desktop.is_absolute() and desktop.is_dir():
                return desktop
    except (OSError, subprocess.SubprocessError):
        pass
    home = Path.home()
    return next((path for path in (home / "Desktop", home / "Bureau") if path.is_dir()), home)

from IPython.display import HTML, display
from PIL import Image


def section_card(number, title, description, children):
    heading = widgets.HTML(value=(
        f'<div class="ms-section-heading"><span class="ms-number">{number}</span>'
        f'<div><h2>{title}</h2><p>{description}</p></div></div>'
    ))
    wrapper = widgets.VBox([heading, *children], layout=widgets.Layout(width="100%"))
    wrapper.add_class("ms-card")
    return wrapper


def field_label(title, description):
    return widgets.HTML(value=(
        '<div class="ms-field-label"><strong>' + html.escape(title) + '</strong>'
        '<span>' + html.escape(description) + '</span></div>'
    ))


field_style = {"description_width": "190px"}
field_width = widgets.Layout(width="550px", max_width="100%")
folder_chooser = FileChooser(str(default_image_directory()), show_only_dirs=True)
folder_chooser.layout = widgets.Layout(width="100%")
folder_chooser.add_class("ms-chooser")

scan_button = widgets.Button(description="Scan image folder", icon="folder-open",
                             layout=widgets.Layout(width="200px", height="39px"))
scan_button.add_class("ms-secondary-button")
scan_summary = widgets.HTML('<span class="ms-muted">Select a directory and scan for .czi, .tif, or .tiff files.</span>')
image_dropdown = widgets.Dropdown(options=[("Scan the folder first", None)],
                                  value=None, description="Preview image",
                                  style=field_style, layout=field_width)
input_card = section_card("01", "Input images",
                          "Choose a local folder. Files are analyzed one-by-one, without tiling.",
                          [field_label("Image folder", "The same folder is used for the batch run."),
                           folder_chooser,
                           widgets.HBox([scan_button, scan_summary],
                                        layout=widgets.Layout(flex_flow="row wrap", align_items="center")),
                           image_dropdown])

channel_widget = widgets.BoundedIntText(value=1, min=1, max=100, description="Myotube channel",
                                        style=field_style, layout=field_width)
scene_widget = widgets.BoundedIntText(value=0, min=0, max=1000, description="Scene (0-based)",
                                      style=field_style, layout=field_width)
time_widget = widgets.BoundedIntText(value=0, min=0, max=10000, description="Time (0-based)",
                                     style=field_style, layout=field_width)
z_widget = widgets.BoundedIntText(value=0, min=0, max=10000, description="Z plane (0-based)",
                                  style=field_style, layout=field_width)
mosaic_widget = widgets.Checkbox(value=True, description="Reconstruct CZI mosaics when supported",
                                 layout=field_width, indent=False)
channels_card = section_card("02", "Image channel and dimensions",
                             "Channel numbering starts at 1, like the original Fiji script; scenes, T and Z start at 0.",
                             [channel_widget, scene_widget, time_widget, z_widget, mosaic_widget])

background_radius_widget = widgets.BoundedIntText(value=200, min=1, max=2000,
    description="Rolling-ball radius", style=field_style, layout=field_width)
background_mean_widget = widgets.FloatText(value=3500, description="Zero-pixel replacement",
                                           style=field_style, layout=field_width)
estimate_background_button = widgets.Button(description="Auto-estimate on preview image", icon="calculator",
    layout=widgets.Layout(width="278px", height="40px"))
estimate_background_button.add_class("ms-secondary-button")
background_estimate_message = widgets.HTML(
    '<div class="ms-note"><strong>One value for all images.</strong> Auto-estimate uses the mean intensity of the 10th–50th percentile of nonzero pixels from the selected image. It is a global approximation of background, not an explicitly selected background-only ROI. Refine the value manually and check the crop. Both values remain fixed throughout the batch.</div>')
threshold_widget = widgets.FloatText(value=200, description="Threshold (raw units)",
                                     style=field_style, layout=field_width)
median_widget = widgets.BoundedIntText(value=1, min=0, max=25, description="Median radius (px)",
                                       style=field_style, layout=field_width)
min_area_widget = widgets.BoundedIntText(value=500, min=1, max=2000000000,
                                         description="Minimum area (px²)",
                                         style=field_style, layout=field_width)
connectivity_widget = widgets.Dropdown(options=[("4-connected (Fiji default)", 1),
                                                ("8-connected", 2)], value=1,
                                       description="Area connectivity", style=field_style,
                                       layout=field_width)
process_card = section_card("03", "Segmentation and quantification",
    "Set one common zero-pixel replacement and one common threshold for every image. You can estimate and refine both on a single representative preview.",
    [background_radius_widget, background_mean_widget,
     widgets.HBox([estimate_background_button],layout=widgets.Layout(width="100%")),
     background_estimate_message, threshold_widget,
     median_widget, min_area_widget, connectivity_widget,
     widgets.HTML(value='<div class="ms-note"><strong>Coverage denominator:</strong> ' 
         'all pixels acquired in the original image (intensity &gt; 0). ' 
         'Original zero pixels are always excluded from the mask and area percentage. ' 
         'Zero replacement affects background subtraction only.</div>')])

crop_size_widget = widgets.Dropdown(options=[("384 × 384",384), ("512 × 512",512),
                                             ("768 × 768",768), ("1024 × 1024",1024)],
    value=768, description="Preview crop size",style=field_style,layout=field_width)
crop_x_widget = widgets.IntSlider(value=50,min=0,max=100,step=5,
    description="Crop X centre (%)",continuous_update=False,style=field_style,
    layout=widgets.Layout(width="650px",max_width="100%"))
crop_y_widget = widgets.IntSlider(value=50,min=0,max=100,step=5,
    description="Crop Y centre (%)",continuous_update=False,style=field_style,
    layout=widgets.Layout(width="650px",max_width="100%"))
prepare_crop_button = widgets.Button(description="Prepare interactive crop",icon="crop",
    layout=widgets.Layout(width="240px",height="44px"))
prepare_crop_button.add_class("ms-secondary-button")
threshold_slider = widgets.FloatSlider(value=200,min=0,max=1000,step=1,
    description="Preview threshold",continuous_update=False,disabled=True,
    readout=True,readout_format=".1f",style=field_style,
    layout=widgets.Layout(width="760px",max_width="100%"))
otsu_button = widgets.Button(description="Suggest Otsu threshold",icon="magic",
    disabled=True,layout=widgets.Layout(width="220px",height="40px"))
otsu_button.add_class("ms-secondary-button")
apply_threshold_button = widgets.Button(description="Apply threshold",icon="check",
    disabled=True,layout=widgets.Layout(width="190px",height="40px"))
apply_threshold_button.add_class("ms-primary-button")
threshold_preview_message = widgets.HTML('<div class="ms-note">Choose a representative location and prepare the crop. Adjusting the slider updates only the crop mask, not the full analysis.</div>')
threshold_crop_state = {"data": None, "signature": None}

def _tuning_image_panel(heading):
    title = widgets.HTML(value=f'<div class="ms-preview-title">{html.escape(heading)}</div>')
    image = widgets.Image(format="png",layout=widgets.Layout(width="100%",max_width="350px"))
    panel = widgets.VBox([title,image],layout=widgets.Layout(width="32%",min_width="190px",flex="1 1 200px"))
    panel.add_class("ms-tuning-panel")
    return panel,image

corrected_panel, corrected_crop_image = _tuning_image_panel("Background-corrected crop")
mask_panel, mask_crop_image = _tuning_image_panel("Thresholded mask")
overlay_panel, overlay_crop_image = _tuning_image_panel("Segmentation overlay")
crop_image_row = widgets.HBox([corrected_panel,mask_panel,overlay_panel],
    layout=widgets.Layout(width="100%",flex_flow="row wrap",gap="12px",display="none"))

preview_card = section_card("04", "Interactive threshold and quality-control preview",
    "A single representative crop helps refine zero replacement and threshold. The values you choose here apply unchanged across the entire batch.",
    [field_label("Interactive crop", "Choose a representative region. Editing Zero-pixel replacement above recomputes the same padded crop automatically; the threshold slider updates its segmentation mask."),
     crop_size_widget, crop_x_widget, crop_y_widget, prepare_crop_button,
     threshold_slider, widgets.HBox([otsu_button,apply_threshold_button],
        layout=widgets.Layout(width="100%",flex_flow="row wrap",gap="10px")),
     threshold_preview_message,crop_image_row])

save_masks_widget = widgets.Checkbox(value=True, description="Save binary masks (0 / 255 TIFF)",
                                     layout=field_width, indent=False)
save_corrected_widget = widgets.Checkbox(value=False, description="Save background-corrected images",
                                         layout=field_width, indent=False)
output_card = section_card("05", "Export options",
    "Creates MyoSurface_Output/ with a unique CSV row per source image, masks, logs and optional corrected images.",
    [save_masks_widget, save_corrected_widget])

fresh_button = widgets.Button(description="Launch new analysis", icon="play",
                              layout=widgets.Layout(width="220px", height="46px"))
resume_button = widgets.Button(description="Resume analysis", icon="repeat",
                               layout=widgets.Layout(width="200px", height="46px"))
fresh_button.add_class("ms-primary-button")
resume_button.add_class("ms-secondary-button")
buttons_box = widgets.HBox([fresh_button, resume_button],
                           layout=widgets.Layout(width="100%", flex_flow="row wrap"))
buttons_box.add_class("ms-buttons")
status_badge = widgets.HTML(
    '<div class="ms-status ms-ready"><span class="ms-status-dot"></span>Ready</div>')
# Same progress presentation as MyoSeg/MyoClass: modern full-width bar,
# current / total count and per-image status list, without a raw output console.
progress = widgets.FloatProgress(value=0, min=0, max=1,
    layout=widgets.Layout(width="100%", height="13px"))
progress.add_class("ms-batch-progress")
progress_label = widgets.HTML(value='<div class="ms-batch-counter">0 / 0 images</div>')
file_status = widgets.HTML(value='<div class="ms-batch-placeholder">File statuses will appear here when the analysis starts.</div>')
batch_summary = widgets.HTML(value="")
batch_progress_panel = widgets.VBox([
    widgets.HTML(value='<div class="ms-batch-heading">Batch progress</div>'),
    progress, progress_label, file_status,
], layout=widgets.Layout(width="100%", min_width="0", gap="8px"))
batch_progress_panel.add_class("ms-batch-panel")
run_card = section_card("06", "Batch analysis",
    "The fixed zero replacement and threshold above are applied to every image. New analysis reprocesses all files; Resume skips unchanged files.",
    [buttons_box, status_badge, batch_progress_panel, batch_summary])

header = widgets.HTML(value=f'''
<div class="ms-hero">
  <div class="ms-hero-top"><span class="ms-mark">MS</span>
  <span class="ms-version">VERSION {APP_VERSION}</span></div>
  <h1>MyoSurface</h1>
  <p>Quantify myotube area coverage from fluorescence microscopy images.
  Automated preprocessing, transparent masks and reproducible exports.</p>
  <div class="ms-hero-foot">
    <span>01 &nbsp;Select images</span>
    <span>02 &nbsp;Preview segmentation</span>
    <span>03 &nbsp;Batch quantify</span>
  </div>
</div>
''')

main_ui = widgets.VBox(
    [header, input_card, channels_card, process_card,
     preview_card, output_card, run_card],
    layout=widgets.Layout(width="100%"))
main_ui.add_class("myosurface-app")


def current_params():
    params = SurfaceParameters(
        channel=channel_widget.value,
        scene=scene_widget.value,
        timepoint=time_widget.value,
        z_plane=z_widget.value,
        reconstruct_mosaic=mosaic_widget.value,
        background_radius=background_radius_widget.value,
        background_mean=background_mean_widget.value,
        threshold=threshold_widget.value,
        min_area=min_area_widget.value,
        median_radius=median_widget.value,
        connectivity=connectivity_widget.value,
        save_masks=save_masks_widget.value,
        save_corrected=save_corrected_widget.value,
    )
    params.validate()
    return params


def current_folder():
    folder = folder_chooser.value
    if not folder or not Path(folder).is_dir():
        raise ValueError("Choose an existing image folder above (click Select).")
    return Path(folder)


def _set_status(text, state):
    status_badge.value = (f'<div class="ms-status ms-{state}">'
                          f'<span class="ms-status-dot"></span>{html.escape(text)}</div>')


def _selected_image_and_params():
    folder = current_folder()
    params = current_params()
    filename = image_dropdown.value
    if not filename:
        raise ValueError("Scan the folder and choose an image first.")
    path = folder / filename
    if path.parent.resolve() != folder.resolve() or not path.is_file():
        raise ValueError("Invalid image selection.")
    return path, params


def _estimate_background(_button=None):
    """Estimate ONCE from the selected preview image, then set a fixed batch value."""
    estimate_background_button.disabled = True
    background_estimate_message.value = '<div class="ms-note">Estimating background from the selected preview image...</div>'
    try:
        path, params = _selected_image_and_params()
        image, _ = load_channel(path, params)
        estimate = estimate_zero_replacement(image)
        # Setting this widget automatically refreshes an existing crop preview.
        background_mean_widget.value = float(estimate["value"])
        background_estimate_message.value = (
            '<div class="ms-note"><strong>' + html.escape(path.name) +
            '</strong> · Suggested fixed batch value: <strong>' +
            f'{estimate["value"]:,.1f}' + '</strong> raw units · ' +
            f'P10: {estimate["p10"]:,.1f} · P50: {estimate["p50"]:,.1f} · ' +
            f'{estimate["positive_samples"]:,} nonzero samples. ' +
            'Manually refine the numeric field while viewing the crop. ' +
            'No value will be re-estimated during the batch.</div>')
        if threshold_crop_state.get("data") is None:
            # Avoid loading this same large image a second time just for the crop.
            _prepare_crop(source_image=image)
    except Exception as error:
        background_estimate_message.value = '<div class="ms-note ms-warning">' + html.escape(str(error)) + '</div>'
    finally:
        estimate_background_button.disabled = False


def _crop_config_signature():
    return (str(folder_chooser.value), str(image_dropdown.value), channel_widget.value,
            scene_widget.value, time_widget.value, z_widget.value, mosaic_widget.value,
            background_radius_widget.value,crop_size_widget.value,
            crop_x_widget.value,crop_y_widget.value)


def _invalidate_crop(change=None):
    threshold_crop_state["data"] = None
    threshold_crop_state["signature"] = None
    threshold_slider.disabled = True
    otsu_button.disabled = True
    apply_threshold_button.disabled = True
    crop_image_row.layout.display = "none"
    threshold_preview_message.value = (
        '<div class="ms-note">Crop position or preprocessing changed. '
        'Click <strong>Prepare interactive crop</strong> to rebuild the preview.</div>')


def _refresh_background_on_edit(change=None):
    """Recompute corrected crop + mask when the shared batch value is edited."""
    item = threshold_crop_state.get("data")
    if item is None:
        return
    if threshold_crop_state.get("signature") != _crop_config_signature():
        _invalidate_crop()
        return
    try:
        params = current_params()
        threshold_preview_message.value = (
            '<div class="ms-note">Updating the preview after zero-pixel replacement change...</div>')
        threshold_crop_state["data"] = refresh_threshold_crop_replacement(item, params)
        _render_threshold_crop()
    except Exception as error:
        # Do not show stale results following a rejected edit.
        _invalidate_crop()
        threshold_preview_message.value = (
            '<div class="ms-note ms-warning">' + html.escape(str(error)) + '</div>')


def _image_bytes(image):
    stream=io.BytesIO()
    Image.fromarray(np.asarray(image,dtype=np.uint8)).save(stream,format="PNG")
    return stream.getvalue()


def _render_threshold_crop(change=None):
    item = threshold_crop_state.get("data")
    if item is None: return
    if threshold_crop_state.get("signature") != _crop_config_signature():
        _invalidate_crop()
        return
    try:
        params = current_params()
        th = float(threshold_slider.value)
        mask = segment_surface_corrected(item["corrected"],item["original"]>0,
                                         params,threshold=th)
        imgs = _make_preview(item["corrected"],mask,max_pixels=850)
        corrected_crop_image.value=_image_bytes(imgs[0])
        mask_crop_image.value=_image_bytes(imgs[1])
        overlay_crop_image.value=_image_bytes(imgs[2])
        crop_image_row.layout.display = "flex"
        nonzero = int(np.count_nonzero(item["original"]>0))
        count = int(np.count_nonzero(mask))
        divisor = nonzero  # Always original acquired ROI (zero pixels excluded).
        value = (100*count/divisor) if divisor else float('nan')
        x0,y0,x1,y1=item["crop_bounds"]
        label = (f'{value:.2f}%' if divisor else 'n/a')
        threshold_preview_message.value = (
            f'<div class="ms-note"><strong>Crop threshold: {th:g} raw units</strong> · '
            f'{count:,} positive px · {label} crop coverage · '
            f'fixed batch zero replacement: {item["replacement"]:,.2f} · '
            f'crop X[{x0}:{x1}], Y[{y0}:{y1}]. '
            'Original zero pixels are excluded from both mask and area denominator. ' 
            'Crop coverage is for visual tuning only; morphology around crop boundaries ' 
            'can differ from the full-image batch result. ' 
            'Click Apply threshold to use this value for the complete batch.</div>')
    except Exception as error:
        threshold_preview_message.value = ('<div class="ms-note ms-warning">' +
            html.escape(str(error)) + '</div>')


def _prepare_crop(_button=None, source_image=None):
    prepare_crop_button.disabled = True
    threshold_preview_message.value = '<div class="ms-note">Reading the image and correcting a padded crop...</div>'
    try:
        path,params = _selected_image_and_params()
        if source_image is None:
            image, _=load_channel(path,params)
        else:
            image=source_image
        data=prepare_threshold_crop(image,params,crop_size_widget.value,
                                     crop_x_widget.value,crop_y_widget.value)
        threshold_crop_state["data"]=data
        threshold_crop_state["signature"]=_crop_config_signature()
        valid=data["corrected"][data["original"]>0]
        if valid.size:
            upper=float(np.percentile(valid,99.8))
        else:
            upper=float(np.max(data["corrected"]))
        target=float(threshold_widget.value)
        slider_max=max(10.,upper*1.2,target*1.25)
        threshold_slider.max=float(slider_max)
        threshold_slider.step=max(.1,float(slider_max)/750.)
        threshold_slider.value=target
        threshold_slider.disabled=False
        otsu_button.disabled=False
        apply_threshold_button.disabled=False
        _render_threshold_crop()
    except Exception as error:
        _invalidate_crop()
        threshold_preview_message.value = '<div class="ms-note ms-warning">' + html.escape(str(error)) + '</div>'
    finally:
        prepare_crop_button.disabled=False


def _update_preview_from_threshold_field(change=None):
    item = threshold_crop_state.get("data")
    if item is None:
        return
    value = float(threshold_widget.value)
    if not np.isfinite(value) or value < 0:
        threshold_preview_message.value = '<div class="ms-note ms-warning">Threshold must be nonnegative and finite.</div>'
        return
    if value > threshold_slider.max:
        threshold_slider.max = max(value*1.25, 10.0)
    threshold_slider.value = value
    _render_threshold_crop()


def _suggest_otsu(_button=None):
    item=threshold_crop_state.get("data")
    if item is None: return
    try:
        from skimage.filters import threshold_otsu
        values=item["corrected"][item["original"]>0]
        if values.size < 16 or float(np.max(values)) <= float(np.min(values)):
            raise ValueError("Insufficient corrected-image contrast for Otsu.")
        value=float(threshold_otsu(values.astype(np.float32)))
        if value>threshold_slider.max: threshold_slider.max=value*1.25
        threshold_slider.value=value
        _render_threshold_crop()
    except Exception as error:
        threshold_preview_message.value = '<div class="ms-note ms-warning">Otsu suggestion unavailable: ' + html.escape(str(error)) + '</div>'


def _apply_preview_threshold(_button=None):
    if threshold_crop_state.get("data") is None or threshold_crop_state.get("signature") != _crop_config_signature():
        threshold_preview_message.value = '<div class="ms-note ms-warning">Regenerate the crop before applying its threshold.</div>'
        return
    threshold_widget.value = float(threshold_slider.value)
    _render_threshold_crop()
    threshold_preview_message.value += '<div class="ms-note ms-applied">Threshold copied to the batch settings. The batch measures each entire acquired image at native resolution.</div>'


def _scan(_button=None):
    """Populate the representative-image menu without a redundant output pane."""
    try:
        images = list_images(current_folder())
        image_dropdown.options = [(p.name, p.name) for p in images] or [
            ("No supported images found", None)]
        scan_summary.value = f'<span class="ms-muted">{len(images)} image(s) found.</span>'
    except Exception as error:
        image_dropdown.options = [("Select a folder first", None)]
        scan_summary.value = '<span class="ms-note ms-warning">' + html.escape(str(error)) + '</span>'


def _make_preview(image, mask, max_pixels=420):
    # Display only: original segmentation is computed at full pixel resolution.
    step = max(1, int(np.ceil(max(image.shape) / max_pixels)))
    image_small = image[::step, ::step]
    mask_small = mask[::step, ::step]
    sample = image_small[image_small > 0]
    if sample.size:
        low, high = np.percentile(sample, [1, 99.5])
        if high <= low:
            low, high = 0, max(1, float(image_small.max()))
    else:
        low, high = 0, 1
    grey = (255 * np.clip((image_small.astype(np.float32) - low) / (high - low),
                          0, 1)).astype(np.uint8)
    original_rgb = np.repeat(grey[:, :, None], 3, axis=-1)
    mask_rgb = np.repeat((mask_small.astype(np.uint8)*255)[:, :, None], 3, axis=-1)
    blend = original_rgb.copy()
    if np.any(mask_small):
        blend[mask_small] = (0.52 * blend[mask_small]
                             + 0.48 * np.array([15, 98, 220])).astype(np.uint8)
        outline = ndi.binary_dilation(mask_small) & ~mask_small
        blend[outline] = (0, 201, 255)
    return [original_rgb, mask_rgb, blend]




# A background worker keeps Voilà responsive and lets ipywidgets render live
# progress as it does in MyoSeg / MyoClass.
_batch_task = None
_batch_files = []
_batch_states = {}
_batch_total = 0
_BATCH_LABELS = {
    "queued": "Pending", "processing": "Processing",
    "completed": "Completed", "skipped": "Skipped", "failed": "Error",
}


def _batch_file_markup():
    if not _batch_files:
        return '<div class="ms-batch-placeholder">No image files queued.</div>'
    # Most recently processed images first, then files not yet visited. This keeps
    # the live status visible even for long batches, and avoids a huge Voilà DOM.
    processed = [name for name in _batch_files if _batch_states[name]["status"] != "queued"]
    waiting = [name for name in _batch_files if _batch_states[name]["status"] == "queued"]
    ordered = processed[::-1] + waiting
    limit = 200
    parts = ['<div class="ms-batch-files" role="list" aria-label="Batch file statuses">']
    for filename in ordered[:limit]:
        details = _batch_states[filename]
        status = details["status"]
        name = html.escape(filename)
        label = _BATCH_LABELS[status]
        coverage = details.get("coverage")
        coverage_text = (f' · {coverage:.3f}%' if isinstance(coverage, (int, float)) else '')
        error_text = details.get("error") or ""
        parts.append(
            '<div class="ms-batch-file" role="listitem">'
            '<span class="ms-batch-file-name" title="' + html.escape(filename, quote=True) + '">' + name + '</span>'
            '<span class="ms-batch-file-state ms-batch-' + status + '">' +
            label + html.escape(coverage_text) + '</span></div>'
        )
        if error_text:
            parts.append('<div class="ms-batch-file-error">' + html.escape(error_text) + '</div>')
    if len(ordered) > limit:
        parts.append(f'<div class="ms-batch-list-limit">Showing {limit} of {len(ordered)} files · Full results are saved in the CSV.</div>')
    parts.append('</div>')
    return ''.join(parts)


def _update_batch_file(filename, status, coverage, error):
    if filename not in _batch_states:
        return
    _batch_states[filename] = {"status": status, "coverage": coverage, "error": error}
    file_status.value = _batch_file_markup()


def _update_batch_progress(current, total):
    progress.value = min(1.0, max(0.0, float(current) / max(1, total)))
    progress_label.value = (f'<div class="ms-batch-counter"><strong>{current} / {total}</strong> images processed'
                            f'<span>{100 * progress.value:.0f}%</span></div>')


def _set_batch_buttons(disabled):
    for control in (fresh_button, resume_button, scan_button,
                    estimate_background_button, prepare_crop_button, otsu_button,
                    apply_threshold_button):
        control.disabled = disabled


async def _run_ui(resume=False):
    global _batch_files, _batch_states, _batch_total
    _set_batch_buttons(True)
    _set_status("Batch analysis running", "running")
    progress.value = 0.0
    progress_label.value = '<div class="ms-batch-counter">Preparing image list…</div>'
    batch_summary.value = ""
    file_status.value = '<div class="ms-batch-placeholder">Scanning the input folder…</div>'
    try:
        folder = current_folder()
        params = current_params()  # Immutable snapshot, shared across entire batch.
        paths = list_images(folder)
        if not paths:
            raise ValueError("No supported .tif, .tiff or .czi files in selected folder.")
        _batch_files = [path.name for path in paths]
        _batch_total = len(paths)
        _batch_states = {name: {"status": "queued", "coverage": None, "error": None}
                         for name in _batch_files}
        _update_batch_progress(0, _batch_total)
        file_status.value = _batch_file_markup()
        loop = asyncio.get_running_loop()

        def on_file(filename, status, coverage, error):
            loop.call_soon_threadsafe(_update_batch_file, filename, status, coverage, error)

        def on_progress(current, total):
            loop.call_soon_threadsafe(_update_batch_progress, current, total)

        result = await loop.run_in_executor(None, lambda: run_batch(
            folder, params, resume=resume, progress_callback=on_progress,
            file_callback=on_file))
        await asyncio.sleep(0)  # Flush queued UI callbacks from the worker thread.
        _update_batch_progress(result["total"], result["total"])
        batch_summary.value = (
            '<div class="ms-batch-result"><strong>' +
            f'{result["processed"]} processed · {result["skipped"]} skipped · {result["failed"]} errors' +
            '</strong><div class="ms-batch-path">CSV: ' + html.escape(result["results_csv"]) +
            '</div><div class="ms-batch-path">Output folder: ' + html.escape(result["output_folder"]) +
            '</div></div>'
        )
        if result["failed"]:
            _set_status("Completed with errors — review file statuses and CSV", "error")
        else:
            _set_status("Batch analysis complete", "complete")
    except Exception as error:
        _set_status("Batch interrupted — see details below", "error")
        batch_summary.value = (
            '<div class="ms-batch-result ms-batch-result-error"><strong>Batch interrupted</strong>'
            '<div class="ms-batch-path">' + html.escape(str(error)) + '</div></div>'
        )
    finally:
        _set_batch_buttons(False)
        # Crop actions are only usable when the preview crop has been prepared.
        ready = threshold_crop_state.get("data") is not None
        otsu_button.disabled = not ready
        apply_threshold_button.disabled = not ready


def _start_batch(resume=False):
    global _batch_task
    if _batch_task is not None and not _batch_task.done():
        return
    try:
        _batch_task = asyncio.get_running_loop().create_task(_run_ui(resume))
    except RuntimeError as error:
        _set_status("Unable to start batch", "error")
        batch_summary.value = '<div class="ms-note ms-warning">' + html.escape(str(error)) + '</div>'


estimate_background_button.on_click(_estimate_background)
prepare_crop_button.on_click(_prepare_crop)
threshold_slider.observe(_render_threshold_crop,names="value")
otsu_button.on_click(_suggest_otsu)
apply_threshold_button.on_click(_apply_preview_threshold)
for _source in (image_dropdown,channel_widget,scene_widget,time_widget,z_widget,
                mosaic_widget,background_radius_widget,
                crop_size_widget,crop_x_widget,crop_y_widget):
    _source.observe(_invalidate_crop,names="value")
background_mean_widget.observe(_refresh_background_on_edit,names="value")
threshold_widget.observe(_update_preview_from_threshold_field,names="value")
for _source in (median_widget,min_area_widget,connectivity_widget):
    _source.observe(_render_threshold_crop,names="value")

scan_button.on_click(_scan)
fresh_button.on_click(lambda _button: _start_batch(resume=False))
resume_button.on_click(lambda _button: _start_batch(resume=True))

display(main_ui)
```
